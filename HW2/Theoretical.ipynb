## 1. Gini Impurity
In class, we defined the Gini impurity as 
$$
\varphi_{Gini}(p) = 1 - \sum_{j=1}^k p_j^2,\qquad p \in [0,1]^k~,
$$
where $p=(p_1,\ldots,p_k)$ represents class proportions in a set of instances. 

1. Prove that 
$$
\varphi_{Gini}(p) \leq 1-1/k.
$$
Hint: find the extermum by solving the vector equation $\nabla \varphi_{Gini}(p)=0$. 

Let $Y_1$ and $Y_2$ be two independent random variables, each represnting the class label of a randomly sampled instance from the set. Namely:
$$
 \Pr[Y_i = j]=p_j, \qquad i\in\{1,2\}, \qquad j\in\{1\ldots k\}~.
$$ 
2. Prove that Gini impurity is the probability that two randomly sampled instances (with replacement) from the set of instances have different class labels. Namley, that
$$
\varphi_{Gini}(p) = \Pr[Y_1 \neq  Y_2].
$$

### Solution 1.1

We shall prove in two ways:

Method 1: Using Cauchy-Schwarz inequality:
Let $\vec{a} = (p_1, p_2, \dots, p_k)$ and let $\vec{b} = (1,1, \dots, 1)$
$$\vec{a} \cdot \vec{b} = \sum_{j=1}^k p_j \cdot 1 = \sum_{j=1}^k p_j = 1$$ 
(since $p$ is a probability distribution)
$$|\vec{a}|^2 = \sum_{j=1}^k p_j^2$$
$$|\vec{b}|^2 = \sum_{j=1}^k 1^2 = k$$  
We substitute it back into the inequality: 
$$1^2 \leq \sum_{j=1}^k p_j^2 \cdot k \Rightarrow 1 \leq k \sum_{j=1}^k p_j^2 \Rightarrow \frac{1}{k} \leq \sum_{j=1}^k p_j^2$$
$$\varphi_{Gini}(p) = 1 - \sum_{j=1}^k p_j^2 \leq 1 - \frac{1}{k}$$


Equality in Cauchy-Schwarz holds when one vector is a scalar multiple of the other: $\vec{a} = \lambda \vec{b}$. This means $p_j = \lambda$ for all $j$.

Since $\sum_{j=1}^k p_j = 1$, we get $k\lambda = 1$, so $\lambda = \frac{1}{k}$.

Method 2: Calculus Approach  
$$\varphi_{Gini}(p) = 1 - \sum_{j=1}^k p_j^2$$  
We know that: $\sum_{j=1}^k p_j = 1$  

$$f(p_1, p_2, ..., p_k, \lambda) = \left(1 - \sum_{j=1}^k p_j^2\right) - \lambda\left(\sum_{j=1}^k p_j - 1\right)$$  

For each $i$ from 1 to $k$: $$\frac{\partial f}{\partial p_i} = -2p_i - \lambda = 0$$

This gives us: $$p_i = -\frac{\lambda}{2}$$

Since all $p_i$ equal the same value $-\frac{\lambda}{2}$, we have $p_1 = p_2 = ... = p_k$.

Therefore: $$p_j = \frac{1}{k}$$ for all $j$

Substituting $p_j = \frac{1}{k}$ into the Gini impurity formula: $$\varphi_{Gini}(p) = 1 - \sum_{j=1}^k p_j^2 = 1 - \sum_{j=1}^k \left(\frac{1}{k}\right)^2 = 1 - k \cdot \frac{1}{k^2} = 1 - \frac{1}{k}$$

### Solution 1.2
$$\Pr[Y_1 \neq  Y_2] = 1 - \Pr[Y_1 = Y_2] = 1 - \sum_{j=1}^k Pr[Y_1=j]\cdot Pr[Y_2=j] = 1 - \sum_{j=1}^k pj^2 = \varphi_{Gini}(p)$$


## 2. Information Gain
In class we claimed that **information gain is always non-negative**. Here, we will prove this for the specific case of binary classification, where we have only two class labels.

Recall that information gain is defined as follows:
$$
IG(S,A) ~~=~~ H(S) - \sum_{v\in Values(A)} \frac{|S_v|}{|S|}H(S_v)~,
$$
where $S$ is a set of data instances, $A$ is an attribute (ferature) with a finite set of possible values $Values(A)$, and $H$ is the entropy function applied to the probability vector associated with the class frequencies. Assuming that there are only two class lables, the entropy can be expressed as follows:
$$
H(S)  ~~=~~ h(p_1) ~~=~~ -p_1\log(p_1)-(1-p_1)\log(1-p_1)~,
$$
where $p_1$ is the frequency of the first label (and $1-p_1$ is the frequency of the second label). Here, we adhere to the convention that $0\cdot \log(0) = 0$ (as $\log(0)$ is undefined). 

We start by examining the function $h()$, which is also called the *binary entropy function* (see plot below). One feature of this function is that it is *concave*. Concave functions satisfy the following property: for every $x_1,x_2\in[0,1]$ and for every $\lambda_1,\lambda_2\in[0,1]$ such that $\lambda_1+\lambda_2=1$, we have:
$$
\begin{align}
h(\lambda_1 x_1 +  \lambda_2 x_2) \geq \lambda_1 h(x_1) + \lambda_2 h(x_2). 
\end{align}
$$

The plot below illustrates this inequality, whose correctness we will assume here without a formal proof (this can be proved by analysis of the first and second derivatives of $h()$).
![Entropy-concave](entropy-concave.png)

1. Use the inequality in (1) to prove (by induction) a more general claim: for any $t\geq 2$ points $x_1\ldots x_t\in [0,1]$, and $t$ weights $\lambda_1\ldots \lambda_t\in [0,1]$ such that $\sum_{j=1}^t\lambda_j = 1$, we have
$$
h\left(\sum_{j=1}^t\lambda_jx_j\right) \geq \sum_{j=1}^t\lambda_jh(x_j)~.
$$
This inequality, which applies to all concave functions, is also called *Jensen's inequality*.

2. Use the inequality you proved above to prove that information gain is always non-negative (when there are only two classes).

### Solution 2.1
#### Base Case: 
Let $t = 2$ then $$\begin{align}h(\lambda_1 x_1 +  \lambda_2 x_2) \geq \lambda_1 h(x_1) + \lambda_2 h(x_2). \end{align}$$
#### Induction Hypothesis:
Assume $ \forall t$ the proposition holds, we shall now prove on $ t+1$  
#### Induction Step: 
let $t+1 \in \N$, then 
$$h\left(\sum_{j=1}^{t+1}\lambda_jx_j\right) = h\left(\sum_{j=1}^t\lambda_jx_j+\lambda_{t+1}x_{t+1}\right) =h\left(\textcolor{green}{\sum_{j=1}^t\lambda_j}\frac{\sum_{j=1}^t \lambda_j x_j}{\sum_{j=1}^t \lambda_j} + \textcolor{green}{\lambda_{t+1}}x_{t+1}\right)$$

note that we meet the condition for the base case of our proof but what is marked $\textcolor{green}{green}$, then we conclude the following:  

$$h\left(\textcolor{green}{\sum_{j=1}^t\lambda_j}\frac{\sum_{j=1}^t \lambda_j x_j}{\sum_{j=1}^t \lambda_j} + \textcolor{green}{\lambda_{t+1}}x_{t+1}\right) \geq h\left(\textcolor{green}{\sum_{j=1}^t\lambda_j}\frac{\sum_{j=1}^t \lambda_j x_j}{\sum_{j=1}^t \lambda_j}\right) + h\left(\textcolor{green}{\lambda_{t+1}}x_{t+1}\right)$$  

By our induction hypothesis we have that:  

$$h\left(\sum_{j=1}^t\lambda_j\frac{\sum_{j=1}^t \lambda_j x_j}{\sum_{j=1}^t \lambda_j}\right) = h\left(\sum_{j=1}^t\lambda_jx_j\right) \geq \sum_{j=1}^t\lambda_jh(x_j)~$$  

Concluding our proof

### Solution 2.2

Information gain is given to us as follows: $$IG(S,A) ~~=~~ H(S) - \sum_{v\in Values(A)} \frac{|S_v|}{|S|}H(S_v)~ $$
We will show that $IG$ is an example of our general claim in 2.1:  

The binary entropy function $h(p) = -p\log(p) - (1-p)\log(1-p)$ is concave, and our weights $\frac{|S_v|}{|S|}$ sum to 1.   
So if $x_v ~=~ p_{1v}$ and $\lambda_v ~=~ \frac{|S_v|}{|S|}$ then: 

$$h\left(\sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot p_{1v}\right) \geq \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot h(p_{1v})$$  

And from the above section we can say:  


$$h(p_1) \geq \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot h(p_{1v})$$


Recall that entropy is defined as:
- $H(S) = h(p_1)$ (entropy of the full dataset)
- $H(S_v) = h(p_{1v})$ (entropy of subset $S_v$)


This gives us:

$$H(S) \geq \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot H(S_v)$$  


By definition, information gain is:

$$IG(S,A) = H(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot H(S_v)$$

From the inequality we just derived:

$$H(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot H(S_v) \geq 0$$

Therefore:

$$IG(S,A) \geq 0$$

This proves that information gain is always non-negative for binary classification.
