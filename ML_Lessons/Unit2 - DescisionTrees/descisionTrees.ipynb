### Descision Tree Construction Algorithm 

1. $\text{Intialize queue Q with root node } v_{root} \text{, asociated with all training samples}$
2. $\text{While } Q \ne \empty \text{ do}$
    - $\text{Pop out v from the Q, and let S be the set of samples that are related to v}$
    - $\text{If all samples in S have the same class label y, then v becomes a leaf and the label is y, continue with the next pop}$
    - $\text{Otherwise, for a perticular attribute A (Categorial Attribute), split S into the possible values (sub-attributes) of A}$
    - $\text{We evaluate the goodness of a split}$
    - $\text{In this case, v will become an internal node, and it will contain the attribute A containing the best split by the evaluation}$
    - $\text{The children of v (which are the result of the best split evaluation) are now pushed in Q}$

### How do we determine what's a good split? (Impurity)

- $\text{Our general objective is to reduce the level of uncertainty/impurity with each time we split}$
- $\text{Intuition: Above we mentioned the notion of "Fully Classified", which in essence implies no levels of uncertainty (completely certain) and completely pure} \\ \text{this is because the split was one sided i.e. over a perticular attribute value there was a conclusive result on the label (either all Yes or all No)}$

- $\text{Let S be a set of training samples} \\ \boxed{\color{dodgerblue}\text{Impurity of S } \varphi(S) \text{:= A function of the class proportions in S (on the labels)}}$

#### Properties that we want 

- $\text{Given we have k possible classes (values od y), } \varphi \text{should satisfy the following intuitive properties:}$
1. $\varphi \text{ is a non-negative function in the relevant range: } D = \{ p \in [0,1]^k | \sum_{j=1}^k = 1 \}$
2. $\varphi \text{is "symmetric" meaning we only care about combination not permutation of the impurity}$
3. $ \large \arg\min_D \varphi(p) = \{p | \exists j \mid p_j = 1\}$
    - $\text{By argMin we mean the LEAST IMPURE (most pure) } \\ \text{so if there exists a single attribute with value of 1 there rest of the attributes will necessarily be 0, by property 1}$
4. $\large \arg\max_D \varphi(p) = \{p | \forall j \mid p_j = \frac{1}{k}\}$
    - $\text{Similarly by argMax we mean the MOST IMPURE} \\ \text{implies all proportions of the labels are distributed uniformly across all attributes}$

$$\text{Example 1:}$$ 
$$\varphi(1,0) < \varphi(0.64,0.36) < \underbrace{\varphi(0.6,0.4) = \varphi(0.4,0.6)}_{Symmetry} < \varphi(0.5, 0.5)$$

#### Impurity Reduction 

$\text{This is a computation applied on the impurity function which will help us determine which path of descisions leads to the fastest/most correct label}$

$$ \boxed{ \Delta \varphi(S,A) = \varphi(S) - \sum_{v \in Values(A)}\frac{S_v}{S}\varphi(S_v) }$$

$\text{Example}$ 

$$\large\Delta \varphi (S, Outlook) = \varphi(Outlook) - \frac{1}{|Outlook|}\left(|Sunny|\varphi(Sunny) + |Overcast|\varphi(Overcast) + |Rain|\varphi(Rain)\right)$$

$$\large\Delta \varphi (S, Outlook) = \varphi(\frac{9}{14}, \frac{5}{14}) - \frac{1}{14}\left( 5\varphi((\frac{2}{5}, \frac{3}{5})) + 4\varphi(\frac{4}{4}, \frac{0}{4}) + 5\varphi(\frac{3}{5}, \frac{2}{5})\right)$$

## Gini Impurity 

$$\large\boxed{\varphi_{Gini}(p) := 1 - \sum^k_{j=1}p_j^2}$$

</center>






#### Example

<center>

$\text{Calculating the Gini entropy Outlook}$

|  $$\text{Applying Gini Entropy}$$|
|-|
|$\varphi_{Gini}(\frac{9}{14}, \frac{5}{14}) = 1 - (\frac{9}{14})^2 + (\frac{5}{14})^2  = 0.46$|
|$\varphi_{Gini}\varphi(\frac{2}{5}, \frac{3}{5}) = 1 - (\frac{2}{5})^2 + (\frac{3}{5})^2 = 0.48$|
|$\varphi_{Gini}(\frac{4}{4}, \frac{0}{4}) = 1 - (\frac{4}{4})^2 + (\frac{0}{4})^2 = 0.0$|
|$\varphi_{Gini}(\frac{3}{5}, \frac{2}{5}) = 1 - (\frac{3}{5})^2 + (\frac{2}{5})^2 = 0.48$|

</center>


$\text{Impurity Reduction}$

$$\large\Delta \varphi (S, Outlook) = \varphi(Outlook) - \frac{1}{|Outlook|}\left(|Sunny|\varphi(Sunny) + |Overcast|\varphi(Overcast) + |Rain|\varphi(Rain)\right)$$


$$\large\Delta \varphi (S, Outlook) = 0.46 - \frac{1}{14}\left(5 \cdot 0.48 + 4 \cdot 0 + 5 \cdot 0.48\right) = \boxed{0.12}$$


<center>

$\text{Calculating the Gini entropy Humidity}$

|  $$\text{Applying Gini Entropy}$$|
|-|
|$\varphi_{Gini}(\frac{9}{14}, \frac{5}{14}) = 1 - (\frac{9}{14})^2 + (\frac{5}{14})^2  = 0.46$|
|$\varphi_{Gini}\varphi(\frac{6}{7}, \frac{1}{7}) = 1 - (\frac{6}{7})^2 + (\frac{1}{7})^2 = 0.24$|
|$\varphi_{Gini}(\frac{3}{7}, \frac{4}{7}) = 1 - (\frac{3}{7})^2 + (\frac{4}{7})^2 = 0.51$|

</center>

$\text{Impurity Reduction}$

$$\large\Delta \varphi (S, Outlook) = \varphi(Outlook) - \frac{1}{|Outlook|}\left(|Sunny|\varphi(Sunny) + |Overcast|\varphi(Overcast) + |Rain|\varphi(Rain)\right)$$


$$\large\Delta \varphi (S, Outlook) = 0.46 - \frac{1}{14}\left(7 \cdot 0.24 + 7 \cdot 0.51\right) = \boxed{0.085}$$

#### Conclusion

$\text{We notice that comparing the impurity reduction between the "Outlook" and the "Humidity", the "Outlook" Attribute will reduce the impurity by a greater amount} \\ \text{So in the algorithm we'll add the children of these nodes in the Queue in the next step of the algorithm}$
