### Descision Tree Construction Algorithm 

1. $\text{Intialize queue Q with root node } v_{root} \text{, asociated with all training samples}$
2. $\text{While } Q \ne \empty \text{ do}$
    - $\text{Pop out v from the Q, and let S be the set of samples that are related to v}$
    - $\text{If all samples in S have the same class label y, then v becomes a leaf and the label is y, continue with the next pop}$
    - $\text{Otherwise, for a perticular attribute A (Categorial Attribute), split S into the possible values (sub-attributes) of A}$
    - $\text{We evaluate the goodness of a split}$
    - $\text{In this case, v will become an internal node, and it will contain the attribute A containing the best split by the evaluation}$
    - $\text{The children of v (which are the result of the best split evaluation) are now pushed in Q}$

---


### How do we determine what's a good split? (Impurity)

- $\text{Our general objective is to reduce the level of uncertainty/impurity with each time we split}$
- $\text{Intuition: Above we mentioned the notion of "Fully Classified", which in essence implies no levels of uncertainty (completely certain) and completely pure} \\ \text{this is because the split was one sided i.e. over a perticular attribute value there was a conclusive result on the label (either all Yes or all No)}$

- $\text{Let S be a set of training samples} \\ \boxed{\color{dodgerblue}\text{Impurity of S } \varphi(S) \text{:= A function of the class proportions in S (on the labels)}}$

#### Properties that we want 

- $\text{Given we have k possible classes (values od y), } \varphi \text{should satisfy the following intuitive properties:}$
1. $\varphi \text{ is a non-negative function in the relevant range: } D = \{ p \in [0,1]^k | \sum_{j=1}^k = 1 \}$
2. $\varphi \text{is "symmetric" meaning we only care about combination not permutation of the impurity}$
3. $ \large \arg\min_D \varphi(p) = \{p | \exists j \mid p_j = 1\}$
    - $\text{By argMin we mean the LEAST IMPURE (most pure) } \\ \text{so if there exists a single attribute with value of 1 there rest of the attributes will necessarily be 0, by property 1}$
4. $\large \arg\max_D \varphi(p) = \{p | \forall j \mid p_j = \frac{1}{k}\}$
    - $\text{Similarly by argMax we mean the MOST IMPURE} \\ \text{implies all proportions of the labels are distributed uniformly across all attributes}$

$$\text{Example 1:}$$ 
$$\varphi(1,0) < \varphi(0.64,0.36) < \underbrace{\varphi(0.6,0.4) = \varphi(0.4,0.6)}_{Symmetry} < \varphi(0.5, 0.5)$$

#### Impurity Reduction 

$\text{This is a computation applied on the impurity function which will help us determine which path of descisions leads to the fastest/most correct label}$

$$ \boxed{ \Delta \varphi(S,A) = \varphi(S) - \sum_{v \in Values(A)}\frac{|S_v|}{|S|}\varphi(S_v) }$$

$\text{Example}$ 

$$\large\Delta \varphi (S, Outlook) = \varphi(Outlook) - \frac{1}{|Outlook|}\left(|Sunny|\varphi(Sunny) + |Overcast|\varphi(Overcast) + |Rain|\varphi(Rain)\right)$$

$$\large\Delta \varphi (S, Outlook) = \varphi(\frac{9}{14}, \frac{5}{14}) - \frac{1}{14}\left( 5\varphi((\frac{2}{5}, \frac{3}{5})) + 4\varphi(\frac{4}{4}, \frac{0}{4}) + 5\varphi(\frac{3}{5}, \frac{2}{5})\right)$$

## Entropy and Information Gain 


<center>

$$\large\boxed{\varphi_{Entropy}(p) := - \sum^k_{j=1}p_j\log(p_j)}$$

</center>

### Confirming our desired properties
1. $ H(p) \ge 0 \text{since } p_j \text{ is between 0 and 1 so } \log(p_j) \text{ is less than 0 and thus it holds}$
2. $ H(1,0, ..., 0) = -\log(1) = 0 \text{ we agree that } 0 \cdot \log(0) = 0 \text {"}$
3. $ H(\frac{1}{k}, ..., \frac{1}{k}) = -k\frac{1}{k}\log(\frac{1}{k}) = \log(k) \text{ (In which this represented the most impure distrubtion of information so we'll prove)}$


<center>

$\boxed{\text{Theorem :} H(p) \le \log(k) \text{ where there's equality } \iff p_1 = \dots = p_k}$

</center>

$\text{We know (proved in calculus) that } \color{lightgreen}\ln(x) \le x -1 $ $ \forall x \gt 0$ <br>

$\text{Then } \ln(k) - H(p) = \\ \ln(k)\sum^k_{j=1}p_j + \sum^k_{j=1}p_j\ln(p_j) \text{ -- Reminder: The sum of proportions equates to 1}$

$ = \sum^k_{j=1}p_j(\ln(k) + \ln(p_j)) = \sum^k_{j=1}p_j\ln(kp_j)$

$ = \sum^k_{j=1}p_j\left(-\ln(\frac{1}{kp_j}) \right) \color{lightgreen}{\ge}$  $\sum^k_{j=1}p_j\left(1-\frac{1}{kp_j}\right) $

$ = \sum^k_{j=1}p_j - \sum^k_{j=1}p_j\frac{1}{kp_j} = 1 - \frac{k}{k} = 0 - \text{simplify and sum over k terms}$

$\rightarrow \ln(k) \ge H(p)$

## Continuous features

$\text{Consider the case where we have a mix of Discrete Attributes and also continous attributes, that using this data we wish to classify our data samples} \\ \text{examples are: Age, Weight, Height, etc}$

#### Given Predefined Domain 

$\text{If we're provided with the information of a way to divide the domain into discrete "bins" then this should be implemented and indeed is a simple solution}$

#### Dynamic Splitting 

$\text{We need to find the best threshold dynamically (unlike above where we're provided one)}$
1. $\text{We sort the values of the feature}$
2. $\text{For each adjacent pari of distinct values, we compute a candidate threshold, by obtaining the average}$
3. $\text{Evaluate each split: for each threshold provided we place the data-set S into the groups}$
4. $\text{We then continue with the algorithm}$
5. $\text{Choose the threshold that maximises the impurity reduction}$

#### Example

$\text{For feature age with values:}$ 

$[20, 25, 32, 40, 45]:$

$\text{Sort:}$

$[20, 25, 32, 40, 45]$

$\text{Candidates:}$

$\large\frac{(20+25)}{2} = 22.5$

$\large\frac{(25+32)}{2} = 28.5$

$\large\frac{(32+40)}{2} = 36$

$\large\frac{(40+45)}{2} = 42.5$

### Problems with Descision Trees 

### Skews towards set with larger partitions

- $\text{We saw two functions that satisfy the impurity properties we desire: }$ 

<center>

$$\large\boxed{\varphi_{Entropy}(p) := - \sum^k_{j=1}p_j\log(p_j)}$$

$$\large\boxed{\varphi_{Gini}(p) := 1 - \sum^k_{j=1}p_j^2}$$

</center>

- $\text{When we try to apply Impurity reduction it was defined as follows: }$

$$ \boxed{ \Delta \varphi(S,A) = \varphi(S) - \sum_{v \in Values(A)}\frac{|S_v|}{|S|}\varphi(S_v) }$$

- $\text{Problem: The issue lies in the equation itself, a partition with many attributes is likely to have hiegher impurity reduction than a partition with few subsets}$
- $\text{As such we need to normalise the calculation}$

$$ \boxed{Gain \ Ratio (S, A) := \frac{IG(S,A)}{SplitInfo(S,A)} \ | \ SplitInfo(S,A) := H(\frac{|S_{att_i}|}{|S|})}$$

## Overfitting
1. $\text{This phenomenon occurs when our Data tree predicts well on the trainning dataset but not on the Validating set or Test set}$
2. $\text{Over large data samples that tree can be very large and memory wise may not be the most efficient}$



#### Solutions - Pre Pruning 

1. $\text{We can solve this issue halting the algorithm early under given conditions}$

    - $\text{We stop pruning at a perticular depth of the tree}$
        - $\text{if we still have impure split at the leaves, we'll take the highest count of a class}$
    - $\text{We'll split and add children if a node is related to a min number of sample cases}$
    - $\text{We'll only split if the impurity is high enough}$
    - $\text{All attributes split the node with information gain below some threshold}$
    - $\text{Only go down another level if the }$ $\color{dodgerblue}\text{Chi-square test}$ $\text{ produces unlikely results}$

#### Solutions - Post Pruning 



#### Chi-Test

- $\text{Assume that we have a set S of samples, with k classes (classifications) which have the following proportions} \\ p_1 \dots p_k$
- $\text{Given a perticular attribute from the set S denoted} S_a \text{ we will label the proportions of these class labels } p'_1 \dots p'_k$

$\text{We then ask given a random subset of S, how likely is it to obtain the classification } \{p'_j\} \text{ from }  S_a \text{ If so then we should prune as it won't provide any new information}$

$\text{To be able to check this likelhod we check } \{|p_1 - p'_1| \dots |p_j - p'_j| \} \text{ and we also check the ratio of the size of the data set to the subset } S_a$

$\text{Formally:} \\ \text{Consider a set of samples, S, with class label proportions } p_1 \dots p_k \text{and a random subset, } S' \subset S, \text{ of size m with class label COUNTS } m_1 \dots m_k \\ \text{Then the folloiwng (random) value}$
 
$$\boxed{\large\chi^2(S') := \sum^k_{j=1} \frac{(m_j-mp_j)^2}{mp_j}}$$ 

$\text{Following a } \chi^2 (chi-squared) \text{ distribution with k - 1 degrees of Freedom (since we have k classes)}$

### Example 

$S'_{counts} = \{2, \ 6, \ 3 \}$

$S'_{proportions} = \{0.2, \ 0.5, \ 0.3 \}$ 

$\large\chi^2 = \frac{(2-11\cdot0.2)^2}{11\cdot0.2} + \frac{(6-11\cdot0.5)^2}{11\cdot0.5} + \frac{(3-11\cdot0.3)^2}{11\cdot0.3} = 0.1$

$\text{We have 3 classes so 2 DoF}$

$\text{Given the Chi-squared table we'd go down the column of DoF and find 2} \\ \text{then try to find the closes two consecutive values on the row} \\ \text{go up to the P-value (probability) to find how likely}$

$\text{In our case we 0.9-0.995 roughly high p-value - This means that this information is likely obtainable with the current data}$

$S'_{counts} = \{200, \ 600, \ 300\}$

$S'_{proportions} = \{0.2, \ 0.5, \ 0.3 \}$ 

$\large\chi^2 = \frac{(200-1100\cdot0.2)^2}{1100\cdot0.2} + \frac{(600-1100\cdot0.5)^2}{1100\cdot0.5} + \frac{(3-1100\cdot0.3)^2}{1100\cdot0.3} = 9.1$

$\text{once again we check the table as described above}$

$\text{In our case we 0.01-0.005 roughly low p-value - This means that this contains useful information}$

### Performance Evaluation

- $\text{We wish to measure how far our prediction function has deviated from the true value (in both validation and Training stage)}$
- $\text{We will apply a summation of a function which will return a value known as } \color{dodgerblue}\text{Empirical Risk}$ $\text{This will in essence tell us how "unreliable/Risky" our predictor (classifier) is}$
- $\text{This method works on most classification perdictors, so will also work on descision trees:}$
$$ L(y, \hat y) = \begin{cases} 1 \quad \text{if } y = \hat y \\ 0 \quad \text{Otherwise} \end{cases}$$
$$\boxed{R(c ; D) = \frac{1}{|D|} \sum_{(x,y)\in D}L(y, c(x))}$$

- $\text{Since classification have an objectively right or wrong prediction (unlike linear regression which is a measure of "how wrong")} \\ \text{we can define accuracy}$ 
$$\boxed{Acc(c ; D) = 1 - R(c;D)}$$

#### Pre-Pruning using Validation Set

1. $\text{Split the data into training and validating set (hold test data aside)}$
2. $\text{Set different values on the pre-pruning parameters}$
    - $\text{For each set of hyperparameters, construct the Descision Tree using the train set.}$
    - $\text{For each Tree created we'll compute the classification accuracy}$
    - $\text{If we're using cross-validation, we'll construct and evaluate every fold and record the average validation accuracy}$
3. $\text{We'll choose the hyper parameters with the highest Validation/Cross Validation Accuracy}$

#### Post-Pruning using Validation Set

1. $\text{Split the data into training and validating set (hold test data aside)}$
2. $\text{Construct a tree using training data}$
    - $\text{It's still possible to apply pre-pruning hyperparameters during construction}$
3. $\text{Prune splits that don't contribute to (increased) validation accuracy:}$
    - $\text{A. For each internal node in tree compute the change (improvement) in validation accuracy obtained by pruning it}$
    - $\text{B. Prune the internal node for which the validation accuracy increase was largest}$
    - $\text{C. Go to A. unless validation accuracy was not significantly increased by the last pruning}$
