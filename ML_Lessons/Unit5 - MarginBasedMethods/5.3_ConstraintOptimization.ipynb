# Optimization with Constraints

$\text{We wish to solve a constraints minimization problem: } \boxed {\min_{x \in \mathbb{R}^d} \{ f(x)\} \mid g_i(x) \le 0 \ \forall i = 1 \dots m}$
- $\text{Context: In our SVM algorithm we wanted: } \boxed{\min_{w \in \mathbb{R}^p} \{ \mid \mid w \mid \mid \} \mid y^{(i)}(w^{\top}x^{(i)} + w_0) -1 \ge 0 \ \forall i = 1 \dots n}$

$\text{Notation:}$
- $\text{We use } x \text{ to denote points in the optimization space } \mathbb{R}^d$
- $ f \text{ is the } objective \ function \text{ and } g_1 \dots g_m \text{ are the } constraints$
- $ \text{If } g_i(x) \le 0 \ \forall i = 1 \dots m, \text{ the } x \text{ is said the be } feasiable$
- $ x^* \in \mathbb{R}^d \text{ is said to be a solution if: } f(x^*) = \min_{x \in \mathbb{R}^d} \{ f(x) \mid g_i(x) \le 0 \ \forall i = 1 \dots m \}$

$\text{Note the following also applies to constrained maximization problems}$


## The Lagrangian Method

$\text{One method to solve this problem is to consider the folllwing function:}$
- $ \mathcal{L}: (\mathbb{R}^d, \mathbb{R}^m) \rightarrow \mathbb{R}$
- $ \mathcal{L}(x, \lambda) = f(x) + \sum^m_{i=1} \lambda_i g_i(x) \mid \lambda_i \text{ are called } \boxed{\text{Largrange Multipliers}}$

---

### Single Equality Constraint

$\text{We wish to solve the following constrained minimization problem: }  \boxed {\min_{x \in \mathbb{R}^d} \{ f(x)\} \mid g(x) = 0 }$

$\text{ We convert it to a lagrangian problem: } \mathcal{L}(x, \lambda) = f(x) + \lambda g(x)$

**Claim** $\text{If } x^* \in \mathbb{R}^d \text{ solves the constrained minimization problem, then } \exist \lambda^* \in \mathbb{R} \mid \nabla \mathcal{L}(x^* , \lambda^*) = 0$
- $\text{This can be helpful since possible solutions can be found by searching points which the gradients of the Lagrangian is 0}$
- $\text{This condition sometimes provides analytical solutions to the constrained optimization problem}$

**Insight** $\text{ Notice that: } \nabla \mathcal{L}(x,\lambda) = 0 \iff \nabla f(x) + \lambda \nabla g(x) = 0 \text{ and } g(x) = 0$
- $\text{We just applied the definition of the lagrangian function and the AOD}$

**Equivelant Claim** $\text{If } x^* \in \mathbb{R}^d \text{ solves the constrainted minimization problem, then } \exist \lambda^* \in \mathbb{R} \mid \nabla f( x^* ) + \lambda \nabla g (x^*) = 0$
- $\text{What this means, is if we find a satisfying } x^* \in \mathbb{R}^d \text{ which } f(x^* ) \text{ is minimal and that } g( x^* ) = 0 \text{ it means that} \\ \text{The vector produced by the derivative of f = the negated vector derivative of g for some multiplier } \lambda$

### Multiple Equality Constraint

$\text{This follows the exact same concept as above just with multiple multipliers and multiple constrained functions}$

$\text{We wish to solve the following constrained minimization problem: }  \boxed {\min_{x \in \mathbb{R}^d} \{ f(x)\} \mid g_i(x) = 0 \ \forall \ i = 1, \dots m}$

**Claim** $\text{If } x^* \in \mathbb{R}^d \text{ solves the constrainted minimization problem, then } \exist \lambda^*  = (\lambda^*_1, \dots, \lambda^*_m) \in \mathbb{R}^m \mid \nabla f( x^* ) + \sum^m_{i = 1}\lambda^*_i \nabla g_i (x^*) = 0 \iff \nabla L(x^*, \lambda^*) = 0$\

**Proof Sketch** 
- $\text{Similar to the above argument}$
- $\text{We observe that } \nabla g_i(x) \ \forall g_i \text{ is perpendicular to the feasible surface}$
- $\text{This means that for an optimal solution } x^* \text{ and a vector u such that } u^{\top}\nabla g_i( x^* ) = 0 \ \forall i =1 \dots m \text{ we have } u^{\top} \nabla f( x^* )=0$
- $\text{This implies that } \nabla f (x^* ) \text{ must be a linear combination of } \nabla g_i ( ^*)$

### Example 1 Minimization constrained with single equality constraint

$ \text{Consider the following constrained problem: } \min_{x \in \mathbb{R}^2} \{x_1^2 + x_2^2 \} \mid x_1 + x_2 = 2$

$\text{How do we solve this?}$

1. $\text{Define objective function f and constrained function g} \\  f(x) = x_1^2 + x_2^2 \qquad g(x) = 2 - x_1 - x_2$
2. $\text{Write the lagrangian Function} \\ \mathcal{L}(x, \lambda) = x_1^2 + x_2^2 + \lambda(2 - x_1 - x_2)$
3. $\text{Solve } \nabla \mathcal{L}(x, \lambda)  = 0  \iff \begin{bmatrix} 2x_1 - \lambda \\ 2x_2 - \lambda \\ 2 - x_1 - x_2 \end{bmatrix} = 0 \implies \frac{\lambda^*}{2} = x_1^* = x_2^* = 1$

### Example 2: Minimization constrained with two Equality constraints
**Problem:**  
Minimize  
$$
f(x_1, x_2, x_3) = x_1^2 + x_2^2 + x_3^2
$$  
subject to:  
$$
g_1(x) = x_1 + x_2 - 1 = 0 \quad \text{and} \quad g_2(x) = x_2 + x_3 - 1 = 0
$$

**Step 1: Define the Lagrangian**  
$$
\mathcal{L}(x_1, x_2, x_3, \lambda_1, \lambda_2) = x_1^2 + x_2^2 + x_3^2 + \lambda_1(x_1 + x_2 - 1) + \lambda_2(x_2 + x_3 - 1)
$$

**Step 2: Compute the Gradient and Set to Zero**  
$$
\nabla \mathcal{L} =
\begin{bmatrix}
2x_1 + \lambda_1 \\
2x_2 + \lambda_1 + \lambda_2 \\
2x_3 + \lambda_2 \\
x_1 + x_2 - 1 \\
x_2 + x_3 - 1
\end{bmatrix}
= 0
$$

This gives the system:
$$
\begin{cases}
2x_1 + \lambda_1 = 0 \\
2x_2 + \lambda_1 + \lambda_2 = 0 \\
2x_3 + \lambda_2 = 0 \\
x_1 + x_2 = 1 \\
x_2 + x_3 = 1
\end{cases}
$$

**Step 3: Express Variables in Terms of Lagrange Multipliers**  
$$
x_1 = -\frac{\lambda_1}{2}, \quad
x_2 = -\frac{\lambda_1 + \lambda_2}{2}, \quad
x_3 = -\frac{\lambda_2}{2}
$$

$\text{Plug into constraints:  From } ( x_1 + x_2 = 1 ):$
$$
-\frac{\lambda_1}{2} - \frac{\lambda_1 + \lambda_2}{2} = 1 \\
\Rightarrow -\lambda_1 - \frac{\lambda_2}{2} = 1 \tag{A}
$$

$\text{From } ( x_2 + x_3 = 1 ):$  
$$
-\frac{\lambda_1 + \lambda_2}{2} - \frac{\lambda_2}{2} = 1 \\
\Rightarrow -\frac{\lambda_1}{2} - \lambda_2 = 1 \tag{B}
$$

Multiply (B) by 2:
$$
-\lambda_1 - 2\lambda_2 = 2 \tag{B'}
$$

Subtract (A) from (B'):
$$
(-\lambda_1 - 2\lambda_2) - (-\lambda_1 - \frac{\lambda_2}{2}) = 2 - 1 \\
\Rightarrow -\frac{3\lambda_2}{2} = 1 \Rightarrow \lambda_2 = -\frac{2}{3}
$$

Plug into (A):  
$$
-\lambda_1 - \left(-\frac{1}{3}\right) = 1 \Rightarrow \lambda_1 = -\frac{2}{3}
$$

**Step 4: Solve for** $( x_1, x_2, x_3 ):$
$$
x_1 = -\frac{\lambda_1}{2} = \frac{1}{3}, \quad
x_2 = -\frac{\lambda_1 + \lambda_2}{2} = \frac{2}{3}, \quad
x_3 = -\frac{\lambda_2}{2} = \frac{1}{3}
$$

**Minimum value:**
$$
f(x^*) = \left(\frac{1}{3}\right)^2 + \left(\frac{2}{3}\right)^2 + \left(\frac{1}{3}\right)^2 = \frac{1}{9} + \frac{4}{9} + \frac{1}{9} = \frac{6}{9} = \frac{2}{3}
$$

### Upper bound on Max Entropy

$\text{Minimize -H(p) = } \sum^n_{i=1}p_ilog(p_i)$

$\text{Subject to } \sum^k_{i=1}p_i -1 = 0$

$\text{Our Lagrangian Function: } \mathcal{L}(p, \lambda) = \sum^n_{i=1}p_ilog(p_i) + \lambda(\sum^k_{i=1}p_i -1)$

$\text{By the Lagrangian Claim if } p* \in \mathbb{R}^k \text{ solves this optimization problem, then there's } \lambda \in \mathbb{R} \quad \forall j = 1 \dots k \mid 0 = \frac{\partial}{\partial p_j} \mathcal{L}(p^*, \lambda)$

$\text{So we solve the partial derivative: }$ $\begin{align} 0 &= \frac{\partial}{\partial p_j} \mathcal{L}(p^*, \lambda) \\ &= \frac{\partial}{\partial p_j}\sum^n_{i=1}p_i\log(p_i) + \frac{\partial}{\partial p_j}\lambda(\sum^k_{i=1}p_i -1) \\ &= \frac{\partial}{\partial p_j} p_j\log(p_j) + \lambda \frac{\partial}{\partial p_j}p_j \\ &= \log(p_j) + \frac{p_j}{p_j} + \lambda \\ &= \log(p_j) + 1 + \lambda \\ &\implies log(p^*_j) = -\lambda^* -1\end{align} $

$ \text{This implies that all } p^*_j \text{ are identical so } p^* = \left (\frac{1}{k}, \dots \frac{1}{k} \right) \text{ and } H(p^*) = -\sum^k_{i=1}p^* \log(p^*_i) = \sum^k_{i=1} \frac{1}{k} \log(k) = \log(k)$

## Min-Max Formulation

$\text{The constrained optimization problem is equivelent to the unconstrained min/max problem: } \min_{x \in \mathbb{R}^d} \{ \max_{\lambda \in \mathbb{R}^m \mid \lambda \ge 0} \{ L(x, \lambda)\} \}$
- $\text{The } \bold{inner \ max } \text{ enforces the constrints via Lagrange mulitpliers}$
- $\text{The } \bold{outer \ min } \text{ is over the original vairables}$

**Proof m=1** 

$\text{If } \lambda_1 = 0 \text{ then we have } f(x) = L(x, \lambda_1) \implies f(x) \le \max_{\lambda_1} \{ L(x, \lambda_1) \}$

$\text{If } \lambda_1 \gt 0 \text{ and } g_1(x) \le 0 \text{ then we have } \lambda_1 g_1(x) \le 0 \implies L(x, \lambda_1) \le f(x) \text{ for any feasible } x \text{ and } \lambda_1 \ge 0 \implies f(x) \ge \max_{\lambda_1 \ge 0} \{L(x, \lambda_1) \}$

**Conclusion** $f(x) = \max_{\lambda_1 \ge 0} \{ L(x, \lambda_1) \} \text{ for any feasible } x \in \mathbb{R}^d$

### Complementary Slackness 

$\text{The constrained optimization problem is equivelent to the unconstriained min/max problem:}$  $$\boxed{ \min_{x \in \mathbb{R}^d}  \{ f(x)\} \mid g_i(x) = 0 \ \forall \ i = 1, \dots m \} = \min_{x \in \mathbb{R}^d} \{ \max_{\lambda \in \mathbb{R}^m \mid \lambda \ge 0} \{ L(x, \lambda)\} \} }$$

**Implications** 

1. $\text{If } x^* \text{ solves the constrained problem, then it also solves the min/max problem with some } \lambda* \in \mathbb{R}^m \mid \lambda ^* \ge 0$
2. $\text{For such } x^* \text{ and } \lambda ^* , \text{ we have } f( x^* ) = L(x^* , \lambda^* ) \implies \sum^m_{i=1} \lambda ^{ * }_i g_i( x^*) = 0$
    - $\lambda_i \gt 0 \implies g_i(x^*) = 0 $
    - $g_i(x^*) < 0 \implies \lambda_i = 0 $


## Dual Optimization problem 

$$\boxed{ \min_{x \in \mathbb{R}^d}  \{ f(x)\} \mid g_i(x) = 0 \ \forall \ i = 1, \dots m \} = \min_{x \in \mathbb{R}^d} \{ \max_{\lambda \in \mathbb{R}^m \mid \lambda \ge 0} \{ L(x, \lambda)\} \} }$$

$\text{In general We know...}$

$$ \min_{u \in A} \{ \max_{v \in B} \{ h(u,v) \} \} \ge \max_{v \in B } \{ \min_{u \in A} \{ h(u,v) \} \}$$

$\text{Which implies that...}$

$$ \min_{x \in \mathbb{R}^d } \{ f(x) \mid g_i(x) \le 0 \ \forall i = 1 \dots m \} = \min_{x \in \mathbb{R}^d} \{ \max_{\lambda \in \mathbb{R}^m \mid \lambda \ge 0} \{ L(x, \lambda)\} \}  \ge \max_{ \lambda \in \mathbb{R}^m \mid \lambda \ge 0} \{ \min_{ x \in \mathbb{R}^d} \{ L(x, \lambda) \} \}$$

$\text{This is helpful since these parameters are typically simpler than the constraints on the ooriginal (primal) variables } x \in \mathbb{R}^d$
- $\boxed{\lambda_i \ge 0, \quad \forall i = 1 \dots m}$  Vs.  $\boxed{g_i(x) \ge 0 \quad \forall i = 1 \dots m}$
- $\text{The dual problem is typically solved by analytically expressing the dual objective:} \quad \boxed{\min_{ x \in \mathbb{R}^d} \{ L(x, \lambda) \}}$
- $\text{Such analytical expression for } \min_{ x \in \mathbb{R}^d} \{ L(x, \lambda) \} \text{ can often be derived by enforcing minimality of } L(x, \lambda) \\ \text{ w.r.t. x via the partial derivative conditions:} \frac{\partial}{\partial x_j} L(x, \lambda) = 0 \quad \forall j = 1, \dots d$

### Example: Dual Optimization with Inequality Constraint

We solve the **primal problem**:

$$
\min_{x \in \mathbb{R}} \left\{ f(x) = (x - 2)^2 \right\} \quad \text{subject to} \quad g(x) = 1 - x \le 0
$$

$\text{Step 1: Define the Lagrangian}$

$$
L(x, \lambda) = (x - 2)^2 + \lambda(1 - x)
$$

$\text{Step 2: Minimize the Lagrangian w.r.t. \( x \)}$

$$
\frac{\partial}{\partial x} L(x, \lambda) = 2(x - 2) - \lambda = 0
$$

$$
x = 2 + \frac{\lambda}{2}
$$

$\text{Step 3: Plug x back into the Lagrangian}$

Compute the **dual function**:

$$
g(x) = L(x, \lambda) = (x - 2)^2 + \lambda(1 - x)
$$

Substitute $x = 2 + \frac{\lambda}{2} $

$$
= \left( \frac{\lambda}{2} \right)^2 + \lambda\left(1 - \left(2 + \frac{\lambda}{2} \right)\right)
$$

Simplify:

$$
= \frac{\lambda^2}{4} + \lambda(-1 - \frac{\lambda}{2}) = \frac{\lambda^2}{4} - \lambda - \frac{\lambda^2}{2}
$$

$$
= -\lambda - \frac{\lambda^2}{4}
$$

$\text{Step 4: Maximize the dual function over \( \lambda \ge 0 \)}$

Maximize:

$$
\max_{\lambda \ge 0} \left\{ g(\lambda) = -\lambda - \frac{\lambda^2}{4} \right\}
$$

This is a downward-opening parabola, so the maximum is at \( \lambda^* = 0 \)

$\text{Step 5: Recover the primal solution}$

From before: 

$ x(\lambda) = 2 + \frac{\lambda}{2} \Rightarrow x^* = 2 $

Check feasibility: 
 
$g(x^*) = 1 - 2 = -1 \le 0 \implies $ Constraint satisfied
