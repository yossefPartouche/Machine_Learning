# Support Vector Machine (SVM)

$\text{Recap: We wanted to find a linear seperation with max-margins, since it would lead to better generalisation}$
- $\text{When we had no constraints this was simple i.e. only based on } \mid \mid w \mid \mid $
- $\text{However, with constraints this proved more difficult which lead us to investigate Dual formulation}$

## Min-Max Formulation of the Max-Margin Problem 

$\text{Original Max-Margin Setup:}$  
$$\min_{w \in \mathbb{R}^p} \left\{ \frac{1}{2} \| w \|^2 \right\} \quad \text{subject to } y^{(i)}(w^\top x^{(i)} + w_0) - 1 \ge 0 \quad \forall i = 1, \dots, n$$

$\text{Rewriting as a constrained optimization problem:}$  
$$\text{Objective function: } f(w, w_0) = \frac{1}{2} \| w \|^2$$  
$$\text{Constraint functions: } g_i(w, w_0) = 1 - y^{(i)}(w^\top x^{(i)} + w_0) \le 0$$  
$$\text{Lagrangian: } \mathcal{L}(w, w_0, \lambda) = \frac{1}{2} \| w \|^2 + \sum_{i=1}^n \lambda_i (1 - y^{(i)}(w^\top x^{(i)} + w_0))$$

$$\text{Min-max formulation: } 
\min_{w, w_0} \left\{ \max_{\lambda \in \mathbb{R}^n, \lambda \ge 0} \mathcal{L}(w, w_0, \lambda) \right\}
\ge 
\max_{\lambda \in \mathbb{R}^n, \lambda \ge 0} \left\{ \min_{w, w_0} \mathcal{L}(w, w_0, \lambda) \right\}$$

$\text{Define: } q_\lambda(w, w_0) := \mathcal{L}(w, w_0, \lambda)$

$$\text{We apply the partial derivative conditions } \frac{\partial}{\partial w_j} q_\lambda(w, w_0) \quad \forall j = 0, \dots, p$$

$$\frac{\partial}{\partial w_j} \left( \frac{1}{2} \| w \|^2 + \sum_{i=1}^n \lambda_i g_i(w, w_0) \right)$$

$$\text{Derivative of the regularizer: } 
\frac{\partial}{\partial w_j} \left( \frac{1}{2} \| w \|^2 \right) = 
\begin{cases}
0 & j = 0 \\
w_j & j > 0
\end{cases}$$

$$\text{Derivative of constraint term: } 
\frac{\partial}{\partial w_j} g_i(w, w_0) = 
\frac{\partial}{\partial w_j} \left( 1 - y^{(i)} w^\top x^{(i)} - y^{(i)} w_0 \right) =
\begin{cases}
- y^{(i)} & j = 0 \\
- y^{(i)} x^{(i)}_j & j > 0
\end{cases}$$

$$\text{Putting together: } 
\frac{\partial}{\partial w_j} q_\lambda(w, w_0) = 
\begin{cases}
- \sum_{i=1}^n \lambda_i y^{(i)} & j = 0 \\
w_j - \sum_{i=1}^n \lambda_i y^{(i)} x^{(i)}_j & j > 0
\end{cases}$$

$$\text{Set derivatives to zero: } \frac{\partial}{\partial w_j} q_\lambda(w, w_0) = 0$$

$$\Rightarrow \nabla q_\lambda(w, w_0) \iff 
\sum_{i=1}^n \lambda_i y^{(i)} = 0 \quad \text{and} \quad 
w = \sum_{i=1}^n \lambda_i y^{(i)} x^{(i)}$$

---

### Derivation of Dual Formulation of SVM

$\text{When } \nabla q_\lambda (w, w_0) = 0$ 

$\text{Simplify:}$ <br> $\begin{align} &=\frac{1}{2}w^{\top}w + \sum^n_{i=1} \lambda_i (1 - y^{(i)}(w^{\top}x^{(i)} + w_0)) \\ &= \frac{1}{2}w^{\top}w +  \sum^n_{i=1} \lambda_i - w^{\top}(\sum^n_{i=1}\lambda_i y^{(i)}x^{(i)}) - w_0\sum^n_{i=1} \lambda_i y^{(i)} \\ &= \frac{1}{2} w^{\top}w + \sum^n_{i=1}\lambda_i - w^{\top}w - w_0 \cdot 0 \\ &= \sum^n_{i=1} \lambda_i - \frac{1}{2} w^{\top}{w} \\ &=\sum^n_{i=1} \lambda_i - \frac{1}{2} \left( \sum^n_{i=1} \lambda_i y^{(i)}x^{(i)}\right)^{\top} \left( \sum^n_{i=1} \lambda_i y^{(i)}x^{(i)}\right) \\ &= \sum^n_{i=1} \lambda_i - \frac{1}{2} \sum^n_{i=1} \sum^n_{j=1} \lambda_i \lambda_j y^{(i)} y^{(j)} (x^{(i)})^{\top} x^{(i)} \end{align}$

$\text{In Conclusion:}$ $$\min_{w, w_0} \mathcal{L}(w, w_0, \lambda) = \min_{w, w_0} q_{\lambda}(w, w_0) = \sum^n_{i=1} \lambda_i -\frac{1}{2}\sum^n_{i=1} \sum^n_{j=1} \lambda_i \lambda_j y^{(i)} y^{(j)} (x^{(i)})^{\top} x^{(i)}$$


## Primal-Dual SVM


$\textbf{1. Primal SVM Problem:}$

$$
\min_{w_0,\, w \in \mathbb{R}^{p+1}} 
\left\{ \frac{1}{2} \| w \|^2 \right\}
\quad \text{subject to} \quad
y^{(i)}(w^\top x^{(i)} + w_0) - 1 \ge 0 
\quad \forall i = 1, \dots, n
$$

$\textbf{2. Dual SVM Problem:}$

$$
\max_{\lambda \in \mathbb{R}^n,\, \lambda \ge 0} 
\left\{ 
\sum_{i=1}^n \lambda_i 
- \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n 
\lambda_i \lambda_j y^{(i)} y^{(j)} 
(x^{(i)})^\top x^{(j)} 
\right\}
\quad \text{subject to} \quad 
\sum_{i=1}^n \lambda_i y^{(i)} = 0
$$

$\text{How do we convert dual varibales to primal variables?}$

- $\text{We mentioned that } w = \sum^n_{i=1}\lambda_iy^{(i)}x^{(i)} \text{(By the requirement } \nabla q_{\lambda}(w, w_0) = 0 )$
- $\text{The complementary slackness of the primal (min-max) problem implies that } \lambda_i > 0 \implies y^{(i)}(w^{\top}x^{(i)} + w_0) -1=0$
- $\text{Complementary slackness implies that the support of the optimla solution consists of } x^{(i)} \text{ that are exactly on the margin}$
- $\text{Most SVM solvers use both formulations to find an optimal solution, and use complementary slackness and other gradient conditions to direct their search}$
- $\text{Solvers use the fact that the support is expected to be small to improve the efficiency of their search}$

## SVM with Kernels and Slack Vairables

The dual SVM: 

$$\max_{\lambda \in \mathbb{R}^n; \lambda \ge 0} \{ \sum^n_{i=1}\lambda_i - \frac{1}{2} \sum^n_{i=1}\sum^n_{j=1} \lambda_i \lambda_j y^{(i)} y^{(j)} (x^{(i)})^{\top} x^{(j)}\} \mid \sum^n_{i=1} \lambda_i y^{(i)} = 0$$

The dual SVM with Kernels to explore non-linear descision boundaries: $$\max_{\lambda \in \mathbb{R}^n; \lambda \ge 0} \{ \sum^n_{i=1}\lambda_i - \frac{1}{2} \sum^n_{i=1}\sum^n_{j=1} \lambda_i \lambda_j y^{(i)} y^{(j)} K(x^{(i)}, x^{(j)}) \} \mid \sum^n_{i=1} \lambda_i y^{(i)} = 0$$

- $\text{We find a max-margin linear seperator in the ambiant space } \mathbb{R}^N$
- $\text{The linear separator in the ambient space corresponds to non-linear descision boundaries (with margin) in the original feature space } \mathbb{R}^p$
- $\text{The number of support vectors, now depends on the dimension of the ambiant space (can be very large)}$
- $\text{We can use any of the kernel functions mentioned in dual perceptron algorithm} \{ Linear, Polynomial, rbf, sigmoid\}$

### Allowing Violations of Margin Constraints

$\text{The original  (Primal) set up of the max-margin problem}$  $$\min_{w \in \mathbb{R}^p} \{ \frac{1}{2} \mid \mid w \mid \mid ^2 \} \quad y^{(i)}(w^{\top}x^{(i)} + w_0) - 1 \ge 0 \quad \forall i = 1 \dots n$$

- $\text{We often wish to allow for some violation of the margin constraints}$
- $\text{This can be done by introducing a } slack \ variables \  \xi_i \text{ that measure the amount of margin violation for sample } x^{(i)}$
$$ \xi_i = \max \{0; 1 - y^{(i)}(w^{\top}x^{(i)} + w_0) \}$$

- $\text{if } \xi_i = 0 \text{ then } x^{(i)} \text{ satisfies the margin constraints}$
- $\text{if } \xi_i > 0 \text{ then } x^{(i)} \text{ violates the margin constrains}$
- $\text{if } \xi_i < 0 \text{ then } x^{(i)} \text{ is misclassified }$

$\text{We then add the slack variables to the objective function and constraints: }$ 

$$
\min_{\substack{w_0,\, w \in \mathbb{R}^{p+1} \\ \xi \in \mathbb{R}^n,\, \xi \ge 0}} 
\left\{ 
\frac{1}{2} \| w \|^2 + C \sum_{i=1}^n \xi_i 
\ \middle| \ 
y^{(i)}(w^\top x^{(i)} + w_0) \ge 1 - \xi_i \quad \forall i = 1, \dots, n 
\right\}
$$

$\text{The Dual Form of SVM with slack variables}$ 

$$
\max_{\substack{\lambda \in \mathbb{R}^n}}
\left\{
    \sum^n_{i=1}\lambda_i - \frac{1}{2}\sum^n_{i=1}\sum^n_{j=1} \lambda_i \lambda_j y^{(i)} y^{(j)} (x^{(i)})^{\top} x^{(j)} \ \middle| \ \sum^n_{i=1} \lambda_i y^{(i)} = 0 \text{ AND } \lambda_i \in [0, C]
\right\}
$$

- $\text{If } \lambda_i = 0 \text{ then } x^{(i)} \text{ satisfies the margin constraints and is not in the support}$
- $\text{If } \lambda_i = C \text{ then } x^{(i)} \text{ violates the margin constraints and is in the support}$
- $\text{If } 0 < \lambda_i < C \text{ then } x^{(i)} \text{ is exactly on the margin and is in the support}$
