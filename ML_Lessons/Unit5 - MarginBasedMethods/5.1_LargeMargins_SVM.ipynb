# Large Margin Classification 

- $\text{Recap on Linear Classification:}$
    - $\text{We used the perceptron algorithm to find a separating hyperplane, when the data is linearly seperable}$
    - $\text{When the data wasn't linearly seperable, we apply the dual perceptron with a kernel function find a seperating hyperplane in a (typically) large ambiant space}$

- $\text{Two main issues:}$
    1. $\text{When the data isn't linearly seperable (in the ambient space), it will not produce a useful classifier}$
    2. $\text{When data is linearly seperable, it might produce a seperating hyperplane that will not generalize well}$
---

## Classification Margins

$\boxed{\text{Margin of a Classifier := } \min_{x \in D} \{ dist(x, \partial M(c)) \}} \\ \text{ where c is the linear classifier and } \partial M(c) \text{ is a separating hyperplane}$

$\text{Large margins are associated with better generalization, since they will accurately classify if test data resembles training data} \\ \text{ 
    - Allows more space for outlier points as well}$

$\text{GOAL: develope an optimization scheme that finds the seperating hyperplane with maximum margins}$

---

### Distance From Hyperplane 

- $\text{The distance between a data instance } x \in \mathbb R^p \text{ and the hyperplane, is defined that by the distance of the vector from the instance and the shortest point on the hyperplane:}$

$$ \text{dist} = \min_{x' \mid w_0 + w^{\top}x' = 0} \{ \mid \mid x - x' \mid \mid \}$$

- $\text{As we can use the implicit form of the hyperplane to classify how far our instance is from the plane, we just need to normalize it by the vector w and so our distance is:}$ $$\boxed{dist = \frac{\mid w^{\top}x + w_0 \mid}{ \mid \mid w \mid \mid}}$$

**Proof** 

$\text{Consider a point } x' \text{ on the hyperplane where } \mid \mid x - x' \mid \mid = \text{dist}$

$$ x - x' = \alpha w $$

$\text{Then }$  $$\mid \mid x - x' \mid \mid = \mid \mid \alpha w \mid \mid = \mid \alpha \mid \ \mid \mid w \mid \mid $$

$\text{Then }$ $$x = x' + \alpha w \implies w^{\top}x = w^{\top}(\alpha w + x') =  \alpha w^{\top}w + w^{\top}x' = \alpha \mid \mid w \mid \mid ^2 - w_0$$

$\text{Combining:}$ $$ \text{dist} = \mid \alpha \mid \mid \mid w \mid \mid = \frac{\mid \alpha \mid \ \mid \mid w \mid \mid ^2 - w_0 + w_0 \mid}{\mid \mid w \mid \mid} = \frac{w^{\top}x + w_0}{\mid \mid w \mid \mid}$$

### Margins of A Linear Classifier

$\text{To represent a hyperplane we have } w_0 \in \mathbb{R} \text{ and } w \in \mathbb{R}^p \text{Will have a classification margin of } \delta \text{ on data D = } \{(x^{(i)}, y^{(i)})\}^n_{i=1} \iff :$

- $y^{(i)} = 1 \implies w^{\top}x^{(i)} + w_0 \ge \delta \mid \mid w \mid \mid $
- $ y^{(i)} = -1 \implies w^{\top}x^{(i)} + w_0 \le \delta \mid \mid w \mid \mid $
- $\text{otherwise written as: }$ $$y^{(i)}(w^{\top}x^{(i)} + w_0) - \delta \mid \mid w \mid \mid \ge 0 \quad \forall i = 1 \dots n$$

$\text{This leads to the }  \bold{Max-Margin \ problem} \text{ := Find the maximum } \delta \in \mathbb{R}^+ \text{ for which there are } w_0 \in \mathbb{R}, w \in \mathbb{R}^p \text{which satisfy the respective inequality for each sample point}$
    - $\text{In essence this is saying to find the maximum distance for each correctly classified point}$

$\text{We wish to consider all } w_0 \in \mathbb{R}, w \in \mathbb{R}^p \text{ that satisfy these inequalities:}$ $$ y^{(i)}(w^{\top}x^{(i)} + w_0) -1 \ge 0 \ \forall i = 1 \cdots n$$
- $\text{Among them we wish to select the solution that minimizes } \mid \mid w \mid \mid (= \frac{1}{\delta})$

$\text{Formally this is known as a type of } \bold{Constrained \ minimzation \ problem}$ $$ \min_{w \in \mathbb{R}^p} \{ \mid \mid w \mid \mid \} \mid y^{(i)}(w^{\top}x^{(i)} + w_0) -1 \ge 0 \ \forall i = 1 \dots n $$

## The Hinge Loss

- $\text{This is a loss function on the instances relative to the margin}$
- $\text{If a sample satisfies the margin, it doesn't contribute to the loss:}$ $$ y^{(i)}(w^{\top}x^{(i)} + w_0) \ge 1 \Longrightarrow L^{(i)}_{hinge} = 0$$

- $\text{if a sample doesn't satisfy the margin, then its loss is proportional to its distance from the margin:}$ $$ y^{(i)}(w^{\top}x^{(i)} + w_0) \lt 1 \Longrightarrow L^{(i)}_{hinge} = 1 - y^{(i)}(w^{\top}x^{(i)} + w_0) $$

- IMAGES

$$\boxed{L^{(i)}_{hinge} = \max \{0, \quad 1 -y^{(i)}(w^{\top}x^{(i)} + w_0)\}}$$

$\text{The total loss is computed as:}$ 

$$L(w, w_0; D) = \frac{1}{2}\mid\mid w \mid \mid ^2 + \frac{C}{n}\sum^n_{i=1}L^{(i)}_{hinge}$$

- $\text{Note we will be changing our hyperplane to find the one with that largest margin, so we want to promote hyperplanes with large margin width and this is represented by:}$ $$\frac{1}{\mid \mid w \mid \mid} \propto \frac{1}{2} \mid \mid w \mid \mid^2$$

- $\text{The constant C is a hyperparameter, which controls the relative effect of margin violations on the hinge loss}$ 
    - $ \text{Smaller C} \implies \text{ Wider margin and more violations}$
    - $ \text{Larger C} \implies \text{ Narrower margins and fewer violations}$

$$ \text{We want: } \min \{ L(w, w_0; D) \} $$

---

### Gradient of Loss

$$L(w, w_0; D) = \frac{1}{2}\mid\mid w \mid \mid ^2 + \frac{C}{n}\sum^n_{i=1}\max \{0, \quad 1 -y^{(i)}(w^{\top}x^{(i)} + w_0)\}$$

$ \frac{\partial L(w, w_0; D)}{ \partial w_j} = \frac{1}{2} \frac{\partial}{ \partial w_j} \color{dodgerblue}{(\mid \mid w \mid \mid)^2 }$ $ + \frac{C}{n}\sum^n_{i=1} \frac{\partial}{\partial w_j}\color{green}\max \{0, \quad 1 -y^{(i)}(w^{\top}x^{(i)} + w_0)\}$

- $\color{dodgerblue} \frac{\partial}{ \partial w_j} (\mid \mid w \mid \mid)^2 = \frac{\partial}{\partial w_j}\sum^p_{l=1}w^2_l = \begin{cases} 0, \quad j = 0 \\ 2w_j, \quad j > 0 \end{cases}$
- $\color{green}\frac{\partial}{\partial w_j}\max \{0, \quad 1 -y^{(i)}(w^{\top}x^{(i)} + w_0)\} \implies -y^{(i)} \frac{\partial}{\partial w_j}(w^{\top}x^{(i)} + w_0) = -y^{(i)} \frac{\partial}{\partial w_j}(\sum^p_{l=1} w_lx_l^{(i)} + w_0) = \begin{cases} -y^{(i)}, \quad j = 0 \\ -y^{(i)}x^{(i)}_j, \quad j \gt 0  \end{cases} = -y^{(i)}x^{(i)}_j \quad \text{(Assuming we denote ) } x^{(i)}_0 = 1$
    - $\color{green}\frac{\partial}{\partial w_j} L^{(i)}_{hinge} = \begin{cases} -y^{(i)}x^{(i)}, & y^{(i)}(w^{\top}x^{(i)} + w_0) < 1 \text{ out of margin } \\ 0, & y^{(i)}(w^{\top}x^{(i)} + w_0) \gt 1 \text{ within margin } \\ NA, & y^{(i)}(w^{\top}x^{(i)} + w_0) = 1 \text{ on the Margin } \end{cases}$

#### Sub-Gradient

$\nabla L(w, w_0; D) = \frac{1}{2} \nabla (\mid \mid w \mid \mid ^2) + \frac{C}{n} \sum^n_{i=1} \nabla L^{(i)}_{hinge}$
- $\nabla (\mid \mid w \mid \mid ^2) = 2(0, w)$
- $\nabla L^{(i)}_{hinge} = \begin{cases} -y^{(i)}x^{(i)}, & y^{(i)}(w^{\top}x^{(i)} + w_0) < 1 \text{ out of margin } \\ 0, & y^{(i)}(w^{\top}x^{(i)} + w_0) \gt 1 \text{ within margin } \\ NA, & y^{(i)}(w^{\top}x^{(i)} + w_0) = 1 \text{ on the Margin } \end{cases}$

### Sub-Gradient Descent

$$L(w, w_0; D) = \frac{1}{2}\mid\mid w \mid \mid ^2 + \frac{C}{n}\sum^n_{i=1}\max \{0, \quad 1 -y^{(i)}(w^{\top}x^{(i)} + w_0)\}$$

$\text{We minimize this loss using gradient descent:}$
- $\text{Initialize w randomly}$
- $\text{Until convergence, do:}$
    - $\text{Compute the set of } \bold{Instances \ that \ violate \ classification \ margins:}$

        $\quad A = \{ i \mid y^{(i)}(w^{\top}x^{(i)} + w_0 ) < 1 \}$
    - $ w = w - \eta w + \eta \frac{C}{n} \sum_{i \in A}y^{(i)}x^{(i)}$
    - $ w_0 = w_0 + \eta \frac{C}{n} \sum_{i \in A} y^{(i)}$
