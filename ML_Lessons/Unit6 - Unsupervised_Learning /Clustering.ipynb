## Simliarity and Distance Measure

### Similarity 

$\text{Q. How do we measure similarity between samples?}$

$\text{Q. How do we evaluate a given partitioning od the data into clusters}$

$\text{A. We often define } similarity \text{ based on } discrepancy \text{ distance: more distance } \implies \text{ the less similar they are}$


$\text{In measure theory a descrepency measure } d \text{ on a set } \chi \ (d: X \times X \rightarrow [0, ∞]) \text{is said to be a distance metric if:}$
1. $\text{Identity of indescernible: } d(u,v) = 0 \iff u = v$
2. $\text{Symmetry: } d(u,v) = d(v,u)$
3. $\text{Triangle inequality: } d(u,v) \le d(u,w) + d(w,u)$

<center>

| $\text{Measure}$ | $\text{Distance/Discrepancy}$| 
|------------------|------------------------------|
| $\text{Euclidean Norm}$ | $$\mid \mathbf{u - v}\mid_2 := \left( \sum^d_{t=1} (u_t - v_t) \right)^{1/2} $$ |
| $L_p \text{ norm}$ |$$\mid \mathbf{u - v}\mid_p := \left( \sum^d_{t=1} (u_t - v_t)^p  \right)^{1/p} $$| 
| $\text{Manhattan distance}$ | $d_{L_1}(u,v) := \sum^n_{i=1} \mid u_i - v_i \mid$ |
| $L_∞ \text{ distance}$ | $d_{L_∞}(u,v) = \max_{i} \{ \mid u_i - v_i \mid \} $|
| $\text{Cosine}$| $d_{cos}(u,v) := 1 - \cos(\theta(u,v)) \\ \text{where } \cos(\theta(u,v)) := \frac{u \cdot v}{\mid\mid u \mid \mid \mid \mid v \mid \mid}$ | 
| $\text{Angle}$| $d_{angle}(u, v) := \theta(u,v)$| 

</center>

$\text{Note that the measure of Euclidean distance is invariant to rigid body trasnformations such as: translations and rotations}$
$\text{But it is variant to non uniform scaling, this means that the data can look different before and after normalisation with this measure}$

## Region-Growing Clustering

- $\text{Places all points separated by some distance threshold } d_0 \text{ in one cluster}$

**Algorithm**

$\text{While some instances are not assigned to clusters:} \\ \quad 1. \text{Pick an un-clustered element (seed) } x, \text{ and assign it to a new cluster } C_x \\ \quad 2. \text{While there are un-clustered elements } x' \text{ within the range } d(x', C_x) \le d_0 \\ \quad \quad - \text{Assign } x' \text{ to } C_x$

**Data-Structure**

$\text{We can use adjacency list/ adjacency Matrix to store the clusters}$ \\

$\text{The complexity is } O(n^2)$

**Typical Use**

$\text{This is used for denoising data or finding outliers}$

## The K-Means Algorithm 

### Within-Cluster Variation (WCSS)

- $\text{Given a Data-set: } D = \{x_i\}^n_{i=1}$
- $\text{A clustering of } D \text{ is defined using a collection of dsjoint sets formally described by:}$
    - $C_1 \dots C_k \subset \{1, \dots, n\} \quad \color{grey} i \in C_j \iff x_j \text{ is in the jth cluster}$
    - $C_1 \cup \dots \cup c_k = \{1, \dots , n \} \quad \color{grey} \text{The clusters cover the data-set}$
    - $C_j \cap c_l = \empty \ \forall j \ne l \quad \color{grey} \text{clusters are disjoint}$

$\text{GOAL: Find the clusters that minimizes the distances within each clusters}$
- $\text{In doing so we'll be finding the most similar set of elements}$
- $\text{We do this using the Euclidean distances measure}$
$$ \boxed{WCSS(D, \{C_j\}^k_{j=1}) := \sum^k_{j=1} \sum_{i \in C_j} \mid \mid x_i - u_j \mid \mid ^2 \quad ; u_j = \frac{1}{\mid C_j \mid}\sum_{i \in C_j}x_i}$$


- $\text{This equation produces a single value (score) which for each point in a cluster } x_i \text{ we compute difference of the average (mean) of the points in the cluster}$
- $\text{So for each cluster } u_j \text{ is computed once and then apply eucleadian norm on each point in the cluster}$

$\text{We then use this as part of the global K-means algorithm, which ireratively decreses WCSS greedily}$

#### Side Note: Adding Weights

$\text{Note: Suppose there existed an allocated weight factor for each instance, then we'd apply the weights to the points when adjusting our centroid, in the above formula}$

$\text{So the new centers are calculated as follows:} $$ \frac{\sum^n_{i=1}w_ix_i}{\sum^n_{i=1}w_i}$$


### K-Means Algorithm

- $\text{Given a Data-set: } D = \{x_i\}^n_{i=1}$
- $\text{Initialize } \mu_1, \dots \mu_k \text{ (randomly) and initialize } C_1, \dots C_k \text{ as empty sets}$
- $\text{Until termination:}$
    1. $\text{For each } i=1, \dots, n: \text{ assign } i \text{ to the cluster } C^*_j \text{ whose centroid is closest to } x_i:$ 
        - $ j^* = \argmin_{j}\{\mid \mid x_i - \mu_j \mid \mid ^2\}$
    2. $\text{For each } j= 1, \dots , k: \text{ update cluster centroid:}$
        - $\mu_j = \frac{1}{\mid C_j \mid}\sum_{i \in C_j} x_i$
    3. $\text{If centroids remained unchanged, then HALT}$

---

### Properties of K-Means

**Complexity**

- $\text{Each iteration takes } O(nkd) \text{ time for computing cluster assignement updates and } O(nd) \text{ time for updating centroids}$
    - $\text{Each point is checked againts the centroid point, and this is repeated until convergence}$
- $\text{Note that the number of iterations can be very large (exponentially in n) if we have carefully constructed data-set}$
- $\text{Typically the complexity is mainly determined by the number of clusters k, and the choice of the intial centroids}$


**Convergence** 

$\text{Claim: Within-cluster variation (WCSS) decreases in every K-means iteration.}$
- $\text{The value of the WCSS objective function, gets smaller each iteration of the mK-means algorithm}$

$\text{Proof: Redefine WCSS to be also a function of } \mu_1, \dots ,\mu_k \text{ by allowing them to be a non-centroid points}$ $$WCSS(D, \{C_j, \mu_j\}^k_{j=1}) := \sum^k_{j=1} \sum_{i \in C_j} \mid \mid x_i - u_j \mid \mid ^2$$
1. $\text{The first part of the iteration fixes } \mu_1, \dots, \mu_k \text{ and finds } C_1, \dots, C_k \text{ to minimize the above expression of WCSS}$
2. $\text{The second part of the iteration fixes } C_1, \dots, C_k \text{ and finds } \mu_1 \dots, \mu_k \text{ to minimize the above expression of WCSS}$

$\text{So, WCSS doesn't inctrease from one iteration to the next, it'll will strictly decrease if we cahnge cluster assignment}$

### Between-Cluster Sum of Squares (BCSS)

- $\text{Our initial objective was to compute clusters that } \bold{minimize} \text{ distances } \bold{within} \text{ each cluster and } \bold{ maximize} \text{ distance } \bold{betweem} \text{ clusters.}$

- $\text{Above we showed that the k-means converges to a (local) minimum of the WCSS, which measures the distance within each cluster}$
- $\text{We'll also prove that by minimizing the WCSS necessarilty also maximizes the between-cluster sum-of-squares, which measures the distance between clusters}$

$\text{Given } D = \{x_i\}^n_{i=1} \text{Define the TOTAL-data sum of squares} \quad \boxed{TSS(D):= \sum^n_{i=1} \mid \mid x_i -\mu \mid \mid ^2 \quad \mu = \frac{1}{n}\sum^n_{i=1}x_i}$

$\text{Claim: for any clustering } C = \{C_1, \dots, C_k \} \text{ of } D \text{ (with associated cluster centroids } \mu_1 \dots \mu_k ), \text{ we have:}$ 
$$ WCSS(D, C) + BCSS(D,C) = TSS{D} \quad \text{where:}$$

$$ WCSS(D, C) := \sum^k_{j=1} \sum_{i \in C_j} \mid \mid x_i - u_j \mid \mid ^2 \quad BCSS(D, C) := \sum^k_{j=1} \mid C_j \mid \mid \mid \mu_j - \mu \mid \mid ^2$$

$\text{Proof:} \quad TSS(D) = \mid \mid x_i - \mu \mid \mid ^2$  $\begin{align} &= (x_i - \mu)^{\top}(x_i - \mu) \\ &= (x_i - \mu_j + \mu-j - \mu)^{\top}(x_i - \mu_j + \mu-j - \mu) \\ &= \left((x_i - \mu_j)^{\top} + (\mu-j - \mu)^{\top}\right)\left((x_i - \mu_j) + (\mu-j - \mu)\right) \\ &= (x_i - \mu_j)^{\top}(x_i - \mu_j) + (x_i - \mu_j)^{\top}(\mu_j - \mu) + (\mu_j - \mu)^{\top}(x_i - \mu_j)+(\mu_j - \mu)^{\top}(\mu_j-\mu) \\ &= \mid \mid x_i - \mu_j \mid \mid ^2 + 2(\mu_j - \mu)^{\top}(x_i - \mu_j) + \mid \mid \mu_j - \mu \mid \mid^2 \end{align}$

$\text{Now consider the sum of the sequence distances from } \mu \text{ to } x_i \text{ associated with cluster } j$ $\begin{align}\sum_{i\in C_j} \mid \mid x_i - \mu \mid \mid ^ 2 & = \sum_{i\in C_j}\left( \mid \mid x_i - \mu_j \mid \mid ^2 + 2(\mu_j - \mu)^{\top}(x_i - \mu_j) + \mid \mid \mu_j - \mu \mid \mid^2 \right) \\ &= \sum_{i\in C_j} \mid \mid x_i - \mu_j \mid \mid ^ 2 + \sum_{i\in C_j} \mid \mid \mu_j - \mu \mid \mid^2 + 2\sum_{i\in C_j} (\mu_j - \mu)^{\top}(x_i - \mu_j) \\ &= \sum_{i\in C_j} \mid \mid x_i - \mu_j \mid \mid ^2 + \mid C_j \mid \mid \mid \mu_j - \mu \mid \mid ^2 + 2 (\mu_j - \mu)^{\top}\boxed{\sum_{i\in C_j}(x_i - \mu_j)} \end{align}$

$\text{We focus on the boxed sum}$ $\begin{align} \sum_{i\in C_j}(x_i - \mu_j) &= \sum_{i\in C_j} x_i - \sum_{i\in C_j} \mu_j \\ &= \sum_{i\in C_j}x_i - \mid C_j \mid \mu_j \\ &= \sum_{i\in C_j}x_i - \mid C_j \mid \frac{1}{\mid C_j \mid} \sum_{i\in C_j} x_i \\ &= \sum_{i\in C_j}x_i - \sum_{i\in C_j} x_i = 0 \in \mathbb{R}^d \end{align}$

$\text{So back to the original proof:}$ $\begin{align} &\sum_{i\in C_j} \mid \mid x_i - \mu_j \mid \mid ^2 + \mid C_j \mid \mid \mid \mu_j - \mu \mid \mid ^2 + 2 (\mu_j - \mu)^{\top}\boxed{\sum_{i\in C_j}(x_i - \mu_j)} \\ &= \sum_{i\in C_j} \mid \mid x_i - \mu_j \mid \mid ^2 + \mid C_j \mid \mid \mid \mu_j - \mu \mid \mid ^2 + 2 (\mu_j - \mu)^{\top} \cdot 0 \\ &= \sum_{i\in C_j} \mid \mid x_i - \mu_j \mid \mid ^2 + \mid C_j \mid \mid \mid \mu_j - \mu \mid \mid ^2 \\ &= WCSS(D,C) + BCSS(D,C) \quad \Box\end{align}$

### Empty Clusters in K-Means

$\text{K-Means algorithm: (with fix for empty clusters)}$
- $\text{Initialize } \mu_1, \dots, \mu_k \text{ (randomly) and initialize } C_1 \dots C_k \text{ as empty sets}$
- $\text{Repeat until convergence of cluster centroids:}$ 
    1. $\text{For each } i = 1 \dots n \text{: Assing } i \text{ to cluster } C_{j^*} \text{ that minimizes } \mid \mid x_i - \mu_j \mid \mid ^2$
    2. $\text{ For each j = 1 \dots k:}$
        - $\text{if } C_j \ne \empty \text{ then update cluster centroid: } \mu_j = \frac{1}{\mid C_j \mid} \sum_{i \in C_j} x_i$
        - $\text{Otherwise:}$
            1. $\text{Choose an index } i \text{ randomly and set } \mu_j = x_i $
            2. $\text{Discard } C_j \text{ and } \mu_j \text{ and continue algorithm with k - 1 clusters}$
            3. $\text{Select the non-empty cluster } C_l \text{ with the largest WCSS } \left( \sum_{i \in C_j} \mid \mid x_i - \mu_l \mid \mid^2 \right) \\ \text{ and split it randomly to two clusters associated with } C_l \text{and } C_j$

### The Effect of K on WCSS

$\text{When k = 1 then } C = \{C_1\} = \{\{1 \dots n \}\} \  WCSS(D, C) = TSS(D) \color{grey}\  \text{All points connected to a single cluster.}$

$\text{When k = n then } C = \{C_1 \dots C_n \} \{\{1\} \dots \{n\}\} \ WCSS(D, C) = 0 \color{grey}\ \text{Each point is their own cluster.}$

$ \forall \ k \text{ we define } WCSS(D, k) = \min_{C = \{C_1 \dots C_k \}} WCSS(D, C) \text{ then WCSS(D,k) is monotonous decreaseing in k }$

- $\text{This makes it difficult to identify a good value for k}$
- $\text{We try to solve this by finding the point k where: } WCSS(D,k-1) << WCSS(D, k) < WCSS(D, k+1)$
    - $\text{This is known as the elbow of the curve}$

### The Silhouette Coefficient 

- $\text{The silhouetter value } S(i) \text{ of a sample } x_i \in D \text{Measures how much a point "should" be in the current cluster}$

1. $ \text{Average distance of samples withing the same cluster } a(i) = \frac{1}{\mid C_j \mid } \sum_{i' \in C_j} d(x_i, x_{i'})$
2. $\text{Average distance to samples in the next closest cluster} b(i) = \min_{l \ne j} \frac{1}{\mid C_l \mid} \sum_{i' \in C_l}d(x_i, x_{i'})$
3. $\text{Silhouette value: } \large{S(i) = \frac{b(i) - a(i)}{\max \{b(i), a(i)\}}}$

#### Interpretatition of Coefficient 

- $S(i) = 1 \implies x_i \text{ is well-clustered in } C_j$
- $S(i) = 0 \implies x_i \text{ is on the border between } C_j \text{ and a competing cluster}$
- $S(i) = -1 \implies x_i \text{ has a poor fit to its cluster } C_j$

- $\text{The silhouette coefficient (SC) measures the "cohesiveness" of a given clustering solution: } SC = \frac{1}{n}\sum^n_{i=1}S(i)$ 

- $\text{We can select the value of k that leads to the maximum SC}$

## Hierarchial Clustering 

$\text{GOAL: Our objective is to define a hierarchy of clusters included in one another such that:}$
- $\text{Top layer consists of a single cluster containing all n samples}$
- $\text{Bottom layer consists of n singleton clusters - one per sample}$
- $\text{Hierarchy is specified useing a tree called a } dendogram \text{ with clusters corresponding to leaves of subtrees}$

$\text{A key advantage of Hierarchial clustering is that we avoid the need to pre-specify the number of desired clusters (k)}$

$\text{It also allows exploration of data structure at different levels of granularity}$

### Approach 1

$\text{Top-down (divisive): progressively splits cluster into two}$

### Approach 2

$\text{Bottom-up (Agglomerative): progressively merge pairs of clusters}$
- $\text{In each iteration, find a pair of closest clusters and merge it}$
- $\text{Continue until we are left with one cluster containing all samples}$


$\text{Note that both methods require computing distances between pairs of clusters}$
