# Guassian Mixture Model and Expectation Maximization

## Multi-Modal Data 

$\text{In Multi modal data, the data distribution has multiple peaks, cannot be sufficiently modeled using simple distributions (i.e. normal)}$

$\text{What we can do instead is model it as a mixture of mixtures of simpler distributions (e.g normal)}$

## The Guassian Mixture Model (GMM)
$\text{Is a common probabilistic model used for modeling multi-modal multi-dimensional data } x \in \mathbb{R}^d$

$\text{It consists of a mixture of k different multi-dimensional normal distributions (Guassisans)}$
- $\text{A sample from the model is generated in 2 steps:}$
    1. $\text{Sample a mixture component } j \in \{1 \dots k \} \text{ according to a categorical distribution with probability vector } \pi \in \mathbb{R}^k$
    2. $\text{Sample from the j-th Guassian according to } N(\mu_j, \sum_j)$

- $\text{The model parameters are: } \{\pi_j, \mu_j, \sum_j\}^k_{j=1}$
    - $\pi_j \ge 0 \text{ being the probability of selecting the j-th Guassian } (\sum^k_j \pi_j = 1)$
    - $\mu_j \in \mathbb{R}^d, \sum_j \in \mathbb{R}^{d \times d} \text{ are the mean and the covariance matrix of the j-th Guassian}$

- $\text{The probability density function (pdf) for data sample } x \in \mathbb{R}^d:$ $$\begin{align} f_X(x \mid \{\pi_j, \mu_j, \Sigma_j\}^k_{j=1}) &= \sum^k_{j=1} \pi_j N(x ; \mu_j, \Sigma_j) \\ &=  \sum^k_{j=1} \pi_j \frac{1}{\sqrt((2\pi)^d | \Sigma_j |)}^{e^{-\frac{1}{2}}(x-\mu_j)^{\top}\Sigma^{-1}_j(x-\mu_j)}\end{align}$$

#### Example 

$\text{Suppose we've received exam scores in ML from two different Universities A and B}$ 
- $\text{University A: scores are around 60 } \pm 10$
- $\text{University B: scores are around 80 } \pm 5$

$\text{Plotting a single Bell Curve (Guassian), won't represent the data well since we have two peaks, so we use two bell curves}$
$\text{So we have k = 2 - multi-dimensional normal distributions}$

$\text{Suppose we want to determine the probability of picking a perticular exam score, here's what we need to consider:}$

1. $\text{We must choose a component (the school), which is a probabilistic event}$
    - $\text{P(School = A) = 0.7}$
    - $\text{P(School = B) = 0.3}$
         - $\text{note the sum of the component is 1}$
2. $\text{Now we choose an exam based on (1), which follows a normal distribution}$
    - $School \ A \sim N(60, 10^2)$
    - $School \ B \sim N(80, 5^2)$

$\text{Altogether we produce the following parameters} \\ \{\pi_1 = 0.7, \mu_1 =60, \sigma_1 = 10^2\} \\ \{\pi_2 = 0.3, \mu_2 =80, \sigma_2 = 5^2\}$

$\text{Suppose we wanted to compute the probability of choosing score = 80}$

$f_x(80) = \pi_1 \cdot N(80; 60, 10^2) + \pi_2 \cdot N(80; 85, 5^2)$

### GMM Likelihood

$\text{Reminder: } \\ \text{When calculating the likelihood, we're trying to answer the following question} \\ \text{How probable is it that my data X was generated under the model with parameters } \theta.$

$\text{To compute the likelihood in a GMM:}$ $$\begin{align}  L\left(\{\pi_j, \mu_j, \Sigma_j\}^k_{j=1}; D\right) &=  f_X(D \mid \{\pi_j, \mu_j, \Sigma_j\}^k_{j=1}) \\ &= \Pi^n_{i=1} f_X(x_i \mid \{\pi_j, \mu_j, \Sigma_j\}^k_{j=1}) \\ &= \Pi^n_{i=1} \sum^k_{j=1} \pi_j \frac{1}{\sqrt((2\pi)^d | \Sigma_j |)}^{e^{-\frac{1}{2}}(x-\mu_j)^{\top}\Sigma^{-1}_j(x-\mu_j)} \end{align} $$

- $\mathbf{We \ assume \ i.i.d.\ between \ instances}$

$\text{To maximize this likelihood function we'll use an iterative approach known as the Expectation Maximization (EM)}$

## Clustering and Mixture Models

$\text{Reminder about clustering:} \\ \text{Given a set of data } D = \{x_i\}^n_{i=1} \subset \mathbb{R}^d$

$\text{Output: A complete partition of } \{1, \dots, n\} \text{ into disjoint subsets } C_1, \dots , C_k$

$\text{Note: The clusters can be seen as labels (or type of classification) } \{y_i\}^n_{i=1} \text{ where } y_i \in \{1, \dots, k \}$

$\text{We can view the label } y_i \text{ as the mixture component from which sample } x_i \text{ was generated under some mixture model, so the clustering algorithm can be used to learn a mixture model}$

### Uniform Isotropic GMM

- $\text{Consider uniform frequencies for all components: } \pi_j = \frac{1}{k}$
- $\text{Identity convariance matrix for all components: } \Sigma_j = I_d$
- $\text{Each feature has a characteristic mean point: } \mu_j \in \mathbb{R}^d$

$\text{The above condition characterise what is known as Uniform Isotropic:}$ 
- $\text{Uniform: Due to the uniform mixture frequencies}$
- $\text{Isotropic because the features are independant and have identical variance}$

$\text{Given the above characteristics the pdf is:}$ $$ \begin{align} f_X(x \mid \{\mu_j\}^k_{j=1}) &= \sum^k_{j=1} \frac{1}{k} N(x ; \mu_j, \Sigma_j = I_d) \\ &= \sum^k_{j=1} \frac{1}{k\sqrt((2\pi)^d | I_d |)}^{e^{-\frac{1}{2}}(x-\mu_j)^{\top}I^{-1}_d(x-\mu_j)} \\ &= \frac{1}{k\sqrt((2\pi)^d)} \sum^k_{j=1} e^{-\frac{1}{2}\| (x-\mu_j) \| ^2 }\end{align}$$

$\text{The Likelihood is:}$ $$L(\{\mu_j\}^k_{j=1}; D) = \boxed{\large k^{-n}(2\pi)^{\frac{-nd}{2}} \Pi^n_{i=1}\sum^k_{j=1}e^{\frac{1}{2} \|x_i - \mu_j\|^2}}$$

$\text{The above formula although correct it's very difficult to maximize, so we'll include the mixture components (or cluster labels) } \{y_i\}^n_{i=1} \text{ as part of the parameters}$

$$\Large{\tilde L \left ( \{\mu_j\}^k_{j=1}, \{y_i\}^n_{i=1}; D \right) = f_X \left( D, \{y_i \}^n_{i=1} \mid \{\mu_j\}^k_{j=1}\right) = k^{-n}(2\pi)^{\frac{-nd}{2}} \Pi^n_{i=1} e^{\frac{1}{2} \| x_i - \mu_{y_i} \|^2}}$$

### The breakdown

- $\text{Each data sample } x_i \text{ is compared to the mean } \mu_j \text{ such that } y_i = j$
- $\text{Adding } \{y_i\}^n_{i=1} \text{ to the parameters of the liklihood, } \tilde L \text{ becomes a product of terms (without the internal sum) and we can consider its log}$

### Log-Likelihood

$\begin{align} \tilde l \left ( \{\mu_j\}^k_j=1, \{y_i\}^n_{i=1}; D \right) &= \log \left[f_X \left( D, \{y_i \}^n_{i=1} \mid \{\mu_j\}^k_{j=1}\right)\right] \\ &= \log \left [k^{-n}(2\pi)^{\frac{-nd}{2}} \Pi^n_{i=1} e^{\frac{1}{2} \| x_i - \mu_{y_i} \|^2} \right] \\ &= \log[k^{-n}(2\pi)^{\frac{-nd}{2}}] -\frac{1}{2} \sum^n_{i=1} \| x_i\ - \mu_{y_i} \|^2 \\ &=  \log[k^{-n}(2\pi)^{\frac{-nd}{2}}] -\frac{1}{2} \sum^k_{j=1} \sum_{i \mid y_i = j} \| x_i\ - \mu_{y_i} \|^2 \end{align}$

### WCSS with Likelihood

$\text{By the final line of the equation we have two parts of the equation and indeed to maximize the likelihood we must minimize the RHS (double summation part)} $

$\text{This is exactly the formulation of the Within Cluster Sum of Squares (WCSS), and indeed we can minimize this via the K-means clustering algorithm}$

$$ WCSS\left(D; \{\mu_j\}^k_{j=1}, \{y_j\}^n_{i=1}\right):= \sum^k_{j=1} \sum_{i \mid y_i = j} \| x_i\ - \mu_{y_i} \|^2 $$

- $\text{K-means prodcues centroids } \{\mu_j\}^k_{j=1} \text{ and cluster assingments }\{y_j\}^n_{i=1} \text{ that minimize the WCSS and thus maximize } \tilde L \\ \text{Although this may produce good estimates of the GMM model parameters, these models may no be the ones that maximize liklihood} \\ \text{this is because GMM and k-means are solving different optimization problems, which only coincide in a restrictive case}$


### Generalised K-Means

$\text{A generalised verison of K-means for estimating paramerters and cluster labels in a general GMM:}$ 
1. $\text{Initialize parameters } \{\pi_j, \mu_j, \Sigma_j, ^k_{j=1}\} \text{ randomly} $
2. $\text{Repeat until cluster sets are no longer updated}$
    - $ y_i \leftarrow j^* = \argmax_{j} \left\{ \pi_j N(x_i \mid \mu_j, \Sigma_j)\right\}$
    - $\text{For each j = 1 ... k, compute MLEs for parameters } \pi_j, \mu_j, \Sigma_j:$
        - $C_j := \{i \mid y_i = j \}; \quad \pi_j \leftarrow \frac{\mid C_j \mid}{n}; \quad \mu_j \leftarrow \frac{1}{\mid c_j \mid} \Sigma_{i \mid i  C_j} x_i; \quad \Sigma_j \leftarrow \frac{1}{\mid C_j \mid}\Sigma_{i \in C_j}(x_i - \mu_j)(x_i - \mu_j)^{\top}$

- $\text{The algorithm finds "Elliptical clusters" of different weihts, whilst the standard K-means algorithm finds spherical clusters with identical weights}$
- $\text{As standard K-means, it maximizes } \tilde L \left( \{\mu_j\}^k_{j=1}, \{y_i\}^n_{i=1} \right) \text{ and not } L (\{\mu_j \}^k_{j=1} ; D )$

## Expectation Maximization (EM)

### K-Means Revisted (uniform istropic model)

- $\text{For each i = 1 ... n, set } y_i \leftarrow \argmax_j \left\{ N(x_i \mid \mu_j, \Sigma = I_d )\right\}$
- $\text{For each j = 1 ... k, set } \mu_j \leftarrow \frac{1}{\mid \{i \mid y_i = j\} \mid} \Sigma_{i \mid y_i = j}x_i$

$\text{Each iteration updates the parameters } \{\mu_j\}^k_{j=1} \text{ based on the most probable cluster assignment } \{y_i\}^n_{i=1}$

### K-Means (Posterior updates)

$\text{Another approach is to consider all possible cluster assignments weighted based on their posterior probabilities}$
$\text{The upside for this approach is that it produces a "smoother" update and also guarranteed to increase the data liklihood in every iteration}$

- $\text{For each i = 1 ... n and j = 1 ... k compute the } responsibility \text{ of cluster j to sample } x_i$ $$ r(i, j) = \frac{1}{R(i)}N(x_i \mid \mu_j, \Sigma = I_d) ; \quad R(i)  = \sum^k_{l=1}N(x_i \mid \mu_l, \Sigma = I_d)$$
    - $\text{probabilistically defined: } r(i, j) = p(y_i = j \mid x_i, \{\mu_j\}^k_{j=1})$

- $\text{For each j = 1 ... k, compute the total responsibility of cluster j } (n_j) \text{ and update } \mu_j$
    - $n_j = \sum^n_{i=1}r(i,j); \quad \mu_j \leftarrow \frac{\sum^n_{i=1}r(i,j)x_i}{n_j}; \quad \sum^k_{j=1}n_j = n$

- $\text{We update the parameters by applying the liklihood maximization formula on the expected values of } n_j = \mid \{i \mid y_i = j\}\mid \land\ \sum_{i \mid y_i = j}x_i$


$\textbf{EM Algorithm for GMMs}$

1. Initialize parameters $\{\pi_j, \mu_j, \Sigma_j\}$ (randomly).  
2. Update (until convergence of parameters / log-likelihood):  
   1. **E-step:** Compute responsibilities for each $i = 1, \dots, n$, $j = 1, \dots, k$:  

      $$
      r_{ij} = \frac{\pi_j \, \mathcal{N}(x_i \mid \mu_j, \Sigma_j)}{\sum\limits_{\ell=1}^k \pi_\ell \, \mathcal{N}(x_i \mid \mu_\ell, \Sigma_\ell)}
      $$

   2. **M-step:** Update parameters for each $j = 1, \dots, k$:  

      $$
      n_j = \sum_{i=1}^n r_{ij}
      $$

      $$
      \pi_j = \frac{n_j}{n}
      $$

      $$
      \mu_j = \frac{1}{n_j} \sum_{i=1}^n r_{ij} x_i
      $$

      $$
      \Sigma_j = \frac{1}{n_j} \sum_{i=1}^n r_{ij} (x_i - \mu_j)(x_i - \mu_j)^\top
      $$

3. After the algorithm converges, we can also assign each sample $x_i$ to a cluster using Bayesian classification.

### General Framework for EM 
$\text{Model and notations:}$
- $\text{Observed data: }X$
- $\text{Parameters: } \theta$
- $\text{Unobserved (hidden) variables: } Y$

$\textbf{The Complete data likelihood} \text{ has "simple form" and can be maximized using an MLE formula (or simple procedure): }$
$$ \tilde L (\theta, Y; \ X) = f_{XY}(X,Y \mid \theta)$$
$$ \theta^* \leftarrow \argmax_{\theta} \{\tilde L( \theta, Y; X) \}$$

$\textbf{The true data liklihood} \text{ involves summing over all possible assignments to the hidden variables Y,} \\ \text{and so it doesn't have a simple form, with multiple local maxima}$

$$L(\theta; X) = f_X(X \mid \theta) = \sum_y f_{XY}(X, Y = y \mid \theta) = \sum_y \tilde L (\theta, Y = y; X)$$

$\text{The efficient procedure for computing } \theta^* \text{ invovles extracting sufficient statistics from the complete data S(X,Y)}$

$$\theta^* \leftarrow MLE(S(X,Y)) := \argmax_{\theta} \{\tilde L(\theta, Y; X)\} $$

## The EM Algorithm 

1. $\text{Set initial (random) values for model parameters: } \theta$
2. $\text{Iterate until convergence of liklihood } L(\theta; X) \text{ or parameter values } \theta:$
    - $\text{Use current parameter values } (\theta) \text{ to compute posterior probabilities for all possible assignments of the hidden vairables}$
        $$ P(Y =y \mid X, \theta)$$
    - $\text{Use these posterior probabilities to compute } \textbf{expected sufficient statistics:} \quad \textbf{E step}$
        $$ \mathbb{E}_{Y \mid X, \theta}[S(X,Y)] = \sum_y P(Y = y \mid X, \theta)S(X,Y = y)$$
    - $\text{Use expected sufficient statistics to } \textbf{update parameters }n \text{based on the MLE produced for } \tilde L(\theta, Y;X) \quad \textbf{M step}$
    $$ \theta \leftarrow MLE(\mathbb{E}_{Y \mid X, \theta}[S(X,Y)] )$$
3. $\text{This update procedure is } \textbf{guaranteed to increase the true data likelihood } L(\theta; X)$
### Properties of EM

1. $\text{EM is a very popular approach for parameter inference in
complex probabilistic models (that involve hidden variables)}$
2. $\text{Guarantees convergence to local maximum of the true data
likelihood function}$
3. $\text{No need to set a learning rate parameter (like in gradient
descent)}$
4. $\text{Convergence could sometimes be slow and there could be
multiple local maxima}$
5. $\text{Alternative version that sets hidden variables to their most
probable posterior assignment} \\  \text{(like what we do in
generalized K-means) tends to make larger update steps, but
it does not guarantee increase in true data likelihood}$
