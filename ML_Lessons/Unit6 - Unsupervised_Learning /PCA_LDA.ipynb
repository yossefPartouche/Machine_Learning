# Principle and Discriminant Component Analysis

- $\text{PCA is geometric approach to dimensionality reduction}$
- $\text{PCA linearly transforms the data to new axes (Principap components) that:}$
    1. $\text{That are orthogonal (which its interpretation implies mutually uncorrelated)}$
    2. $\text{Maximize variance (spread) of the projected data}$

- $\text{PCA finds a sequence of linear combinations of features that have maximum variance, and are mutually uncorrelated}$

#### How does this help in Unsupervised Learning 
1. $\text{Reveals dominant patterns or trends in high-dimensional data}$
2. $\text{Visualizes high-dimensional datasets in 2D (3D)}$
3. $\text{Reduces the dimensions of the data-set}$
    - $\text{Data compression via projection onto the top k components that convey most of the variance}$
    - $\text{Simplify classifiers and avoiding overfitting (few features)}$
    - $\text{Improve numerical stability}$
4. $\text{Denoises data, noise usually doesn't follow the dominant patterns, so applying PCA, just "mutes" the noise followig the main trends}$

## The $1^{st}$ Principle Component

$\text{Given a Data } D= \{x^{(i)}\}^n_{i=1} \mid x^{(i)} \in \mathbb{R}^d $

$\text{We use matrix representation: } X = \left(X_1, \dots, X_d\right) \in \mathbb{R}^{n \times d} \mid ([X]_{ij} = x^{(i)}_j)$
- $\text{Extracting that i-th instance (sample point) and inspect the j-th attribute/property}$

$\text{Since we focus on variance, we assume that each feature } X_j \text{ has been centered to have mean zero } \sum^n_{i=1}[X_{ij}] = 0 \mid \ \forall j$

### The principle 

- $\text{The first principle component (PC) of } X \text{ is a normalized linear combination of } X_1 \dots X_d (\text{ defined by } \phi_1 \in \mathbb{R}^d \text{ s.t. } \mid \mid \phi_1 \mid \mid = 1) \text{ with maximum variance: }$

$$ Z_1 = \phi_{11}X_1 + \phi_{12}X_2 + \dots + \phi_{1d}X_{d} = X \phi_{1} \in \mathbb{R}^n $$

$$ \max \left\{ \frac{1}{n} \sum^n_{i=1}Z^2_{1i} \right\} = \max \left\{ \frac{1}{n} \sum^n_{i=1} (\phi_1 x^{(i)})^2 \right\} \text{ where }  \phi_1 \text{ is the first PC loading vector}$$

- $\text{We're computing the variance with the First PC loading vector}$

### Optimization Problem

$\text{Indeed we have a constrained optimization problem:}$ 

$$ \max_{\phi_1 \in \mathbb{R}^d} \left\{ \frac{1}{n} \sum^n_{i=1} (\phi_1 x^{(i)})^2 \right\} $$

$\text{Indeed the way we described the linear combination is the solution to the constraint:} \quad \mid \mid \phi_1 \mid \mid  = 1 $




### Geometrically

$\boxed{\text{claim: } \bold{maximum} \text{ variance value, we are } \bold{minimizing} \text{ the mean squared Euclidean distance, of the samples from the line represented by } \phi_1}$

#### Proof

$\text{Let } X \in \mathbb{R}^{n \times d} \text{ be the mean centred data matrix (each row is a sample, as shown above)}$

$\text{Let } \phi_1 \in \mathbb{R}^d, \text{ be a unit vector}$

$\text{Each data point } x^{(i)} \in \mathbb{R}^d \text{ is a projected onto the line spanned by } \phi_1 \text{ by}: \\ \hat{x}^{(i)} = \left(\phi^{T}_1x^{(i)}\right)\phi_1$

$\text{We now present the formula of the Euclidean Squared error in the above terms: } \mid \mid x^{(i)} - \hat x^{(i)} \mid \mid ^2 = \mid \mid x^{(i)} - (\phi^{T}_1x^{(i)})\phi_1 \mid \mid ^2$

$\text{Let's sum this over all samples to get total reconstruction error: } E(\boldsymbol{\phi}_1) = \sum{i=1}^n \| x^{(i)} - (\boldsymbol{\phi}_1^\top x^{(i)}) \boldsymbol{\phi}_1 \|^2$

$\text{We'll use the above son for now...}$

$\text{By the projection identity we have: } \| x \|^2 = \| \text{projection} \|^2 + \| \text{residual} \|^2 \implies \| x \|^2 = \| \hat x^{(i)} \|^2 + \|  x^{(i)} - \hat x^{(i)} \|^2$

$\text{So: }  \|  x^{(i)} - \hat x^{(i)} \|^2 = \| x \|^2 - \boxed{\| \hat x^{(i)} \|^2}$

$\text{By the above we know that: } \| \hat x^{(i)} \|^2 =  \| (\boldsymbol{\phi}_1^\top x^{(i)}) \boldsymbol{\phi}_1 \|^2 = (\boldsymbol{\phi}_1^\top x^{(i)})^2 \cdot \|\boldsymbol{\phi}_1\| = (\boldsymbol{\phi}_1^\top x^{(i)})^2$

$\text{To total sum reconstruction error: } E(\boldsymbol{\phi}1) = \sum{i=1}^n \left\| x^{(i)} - (\boldsymbol{\phi}1^\top x^{(i)}) \boldsymbol{\phi}1 \right\|^2
= \sum{i=1}^n \left( \left\| x^{(i)} \right\|^2 - \left( \boldsymbol{\phi}1^\top x^{(i)} \right)^2 \right)
= \sum{i=1}^n \left\| x^{(i)} \right\|^2 - \sum{i=1}^n \left( \boldsymbol{\phi}_1^\top x^{(i)} \right)^2$

- $\text{Note: While it may look like we’re “distributing the squaring operation”, what’s really happening is an application of the Pythagorean theorem.} \\ \text{Since } \| \boldsymbol{\phi}_1 \| = 1 \Rightarrow \| \boldsymbol{\phi}_1 \|^2 = 1, \text{ the projection }(\boldsymbol{\phi}_1^\top x^{(i)}) \boldsymbol{\phi}_1 \text{ and the residual vector form a right angle. So, for each sample point } x^{(i)}, \\ \text{ we’re essentially building a “right triangle” (in quotes, because this geometric intuition holds even in high-dimensional spaces).}$

$\text{Only the second term depends on } \phi_1 \text{ so minimizing error } \iff \text{ we're maximizing } \sum_{i=1}^n (\boldsymbol{\phi}_1^\top x^{(i)})^2 = \| X \boldsymbol{\phi}_1 \|^2 = \boldsymbol{\phi}_1^\top X^\top X \boldsymbol{\phi}_1$

### Analytically Solving

$\text{This optimization problem can be solved analytically by finding the } \bold{eigenvector} \phi_1 \text{ of the } d \times d \text{ sample covariance matrix } \hat \sum = \frac{1}{n}X^{\top}X \text{ that corresponds to its largest eigenvalue } \lambda_1$

$\text{The eigenvalue } \lambda_1 \text{ is the variance of } Z_1:$

$$\lambda_1 = \sum^n_{i=1} Z^2_{1i} = \sum^n_{i=1}(\phi^{\top}_1x^{(i)})^2$$

## The $2^{nd}$ Principle Component

$\text{Given centered data in matrix form: } X = (X_1, \dots, X_d) \in \mathbb{R}^{n \times d} \text{ s.t. } \sum^n_{i=1}X_{ij} = 0$

$\text{The second principle component of X is a normalized linear combination of } X_1 \dots X_d \text{ that has maximum variance among those that are } \boldsymbol{Uncorrelated} \text{ with } Z_1$

$$\max_{\phi_2 \in \mathbb{R}^d} \sum^n_{i=1} (\phi^{\top}_2 x^{(i)}) \text{ s.t. } \| \phi_2\| = 1 \text{ and } \phi_2 \cdot \phi_1 = 0$$

- $\text{When we apply the condition that } \phi_1 \text{ and } \phi_2 \text{ are orthogonal we ensure that } Z_1 \text{ and } Z_2 \text{ are uncorrelated}$

### Geometrically 

$\text{We have two vectors that are orthonormal to another, and each unit vectors thus creating a 2D linear subspace, where all the sample points are being projected into this sub-space,} \\  \text{while keeping the directions of maximum  (dominant attributes) and second maximum variance, thus retaining as much info. as possible whilst compressing the data}$

## Optimizing the $j^{th}$ Principle Component

$\text{The } j-th \text{ PC loading vector } \phi_j \in \mathbb{R}^d \text{is a solution to the following constrained optimization problem}$ 
$$ \max_{\phi_j \in \mathbb{R}^d} \sum^n_{i=2}\left(\phi^{\top}_j x^{\top} \right)^2 \text{ s.t. } \| \phi_j \| = 1 \text{ AND } \phi^{\top}_j\phi_l = 0, \quad \forall l = 1 \dots j-1$$ 

- $\text{So once we've solved for all 1 to j } \phi \text{'s then we can read the data a bit more clear}$
- $\text{Once more this optimization problem can be solved analytically wby finding the eigenvector } \phi_j \text{ of the } d \times d \text{sample covariance matrix } \hat \sum = \frac{1}{n}X^{\top}X \text{ that corresponds to j-th largest eigenvalue } \lambda_j$

### Proportiona Vairance Explaioned (PVA)

- $\text{To understand the strength of each PC, we measure the rpoportion of variance explained (PVE)}$
- $\text{The total variance present in a data set is defined as: } \sum^d_{j=1}Var[X_j] = \sum^d_{j=1} \frac{1}{n} \sum^n_{i=1}x^2_{ij} \mid X_j \text{ is centered so the Mean = 0}$
- $\text{The variance explained by the j-th PC is } Var[Z_j] = \lambda_j$
- $\text{we have (Total variance preservations)} \sum^d_{j=1}Var[X_j] = \sum^d_{j=1}\lambda_j = \sum^d_{j=1}Var[Z_j]$
- $\text{So the PVE of the j-th PC is: }$ $$\boxed{\frac{Var[Z_j]}{\sum^p_{l=1}\lambda_l} = \frac{\lambda_j}{\sum^p_{l=1}\lambda_l}}$$
    - $\text{This sum to one and can be presented (sometimes is) in cumulative form}$
    - $\text{In many applications, we keep the top K PCs: } \tilde{Z} = [Z_1 \mid Z_2 \mid \dots \mid Z_k ] \in \mathbb{n \times k}$

### PCA in Practice

$\text{We'll want to use PCA when:}$ 
1. $\text{Data has correlated features}$
2. $\text{You want to reduce noise/dimension}$
3. $\text{You want to visualize in 2D/3D}$
4. $\text{You don't interpretability of original features}$

- $\text{Oftern used before classification or Clustering}$
- $\text{Can reduce overfitting}$

$\text{Though it has some limitations:}$
1. $\text{Only finds linear structure}$
2. $\text{May discard usefull components}$

## Linear Discriminant Analysis (LDA)

$\text{In Linear Discriminant Analysis (LDA) we seek a vector } \phi_1 \in \mathbb{R}^p \text{ that solves the following: }$ $$\max_{\phi} \frac{BCSS(\phi)}{WCSS(\phi)} \quad (*)$$

$ \text{ where } WCSS(\phi) := \sum^k_{j=1} \sum_{i \in C_j} \|\phi^{\top}x^{(i)} - \phi^{\top}\mu_j \|^2 $

$ \text{ And } BCSS(\phi) := \sum^k_{j=1} | C_j | \| \phi^{\top} \mu_j - \phi^{\top}\mu \| ^2 $

$\mu_j = \frac{1}{| C_j |} \sum_{i \in C_j}x_i \quad$
$\mu = \frac{1}{n} \sum^n_{i=1}x_i $

$\text{Similar to PCA we have multiple Linear Discriminant directions:}$ 
- $\text{The solution } \phi_1 \text{ is called the first Linear Discriminant Direction}$
- $\text{The second Linear Discriminant Direction } \phi_2 \text{ maximizes } (*) \text{ subject to } \phi^{\top}\phi_1 = 0 $
- $\text{The j-th Linear Discriminant Direction } \phi_j \text{ maximizes } (*) \text{ subject to } \phi^{\top}\phi_1 = 0, \dots , \phi^{\top}\phi_{j-1} = 0$
