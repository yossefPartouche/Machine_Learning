# The Kernel Perceptron 

## Motivation For Kernel Function

### Higher Dimension Mapping - Recap

| $\text{Input}$ | $\text{Map}$ | $\text{Algorithm}$ | 
|----------------|--------------|--------------------|
| $\text{D = } \{x^{(i)}, y^{(i)}\}^n_{i=1}, \text{ hyperparameter } \eta > 0$ | $\text{For  i = 1 ... n; } \bar{x}^{(i)} = \varphi(x^{(i)}) \in \mathbb{R}^N$| $\bold{Initialize: } \bar w \in \mathbb{R}^N \text{randomly / arbituarily} \\ \bold{iterate: } \text{until termination} \\ \quad \text{For i = 1, ..., n:} \\ \quad \quad \bold{predict: } \ \hat{y}^{(i)} = sign(\bar{w}^{\top} \bar{x}^{(i)}) \\ \quad \quad \bold{Update: } \ \bar{w} = \begin{cases} \bar{w} - 2 \eta \bar x & \mid y^{(i)} = -1 \ \& \  \hat y^{(i)} = 1 \\ \bar{w}+ 2 \eta \bar x & \mid y^{(i)} = 1 \ \& \  \hat y^{(i)} = -1 \end{cases}$|

$\text{Problem: The ambiant dimension } (N) \text{ can be very large which will dramatically increase the complexity of the algorithm}$

### The Kernel Idea

$\text{We want to create a set of operations which would emulate the Ambiant Space (The linearly seperable space), whilst only using the operations of the original features space}$

## The Dual Perceptron

- $\text{First: We can compactly express the equation } \\ \quad w = w + 2\eta y^{(i)} x^{(i)} \mid y^{(i)} = \ sign(w^{\top}x^{(i)})$
- $\text{Second: If we initialize the algorithm with w = 0 then the algorithm can be represented by the following linear combination} \\ \quad w = \sum^n_{i=1} a_iy^{(i)}x^{(i)} \\ \quad \text{This helps since we don't need to store w in the feature space } \mathbb{R}^p, \text{ we'll hold our a in the } \mathbb{R}^N \text{space,} \\ \quad \text{which will contain w as well.}$

### Dual Perceptron Algorithm 

- $\bold{Initialise \ a = 0} \in \mathbb{R}^n \\ \bold{Iterate} \text{ until termination} \\ \quad \text{For i = 1, ..., n:} \\ \quad \bold{Predict:} \ \hat{y}^{(i)} = sign \left(\sum^n_{j=1}a_jy^{(j)}\left(x^{(j)}\cdot x^{(i)} \right)  \right)\\ \quad \bold{Update: } \text{if } (y^{(i)} \ne \hat y^{(i)}) \text{ then } a_j = a_i + 2\eta $

#### Breakdown of what we changed 

$\textbf{We are applying linearity and the properties of the transpose:}$

$\text{Given:}$

$w = \sum_{j=1}^n a_j y^{(j)} x^{(j)}$

$\text{Then:} \\ 
\begin{aligned}
w^\top x^{(i)} 
&= \left( \sum_{j=1}^n a_j y^{(j)} x^{(j)} \right)^\top x^{(i)} \\
&= \sum_{j=1}^n \left( a_j y^{(j)} x^{(j)} \right)^\top x^{(i)} \\
&= \sum_{j=1}^n a_j y^{(j)} (x^{(j)})^\top x^{(i)}
\end{aligned} $

## Kernel Function - Definition

$\boxed{ \text{Let K :} \mathbb{R}^p \times \mathbb{R}^p \rightarrow \mathbb{R} \text{ be a mappping for the function } \varphi : \mathbb{R}^p \rightarrow \mathbb{R}^N \\ \text{ K is said to be a Kernel Function } \iff K(u, v) = \varphi(u)^{\top}\varphi(v) \ \forall \ u, v \in \mathbb{R}^p}$

### Example 1 - full Quadratic Variety 

- $\text{Suppose we wish to cosider the Full Quadratic Variety for } x \in \mathbb{R}^p$
- $\text{We achieve this as follows: } \varphi(x) = \left[ 1,\ x_1,\ x_2,\ \dots,\ x_p,\ x_1^2,\ x_1x_2,\ x_1x_3,\ \dots,\ x_1x_p,\ x_2^2,\ x_2x_3,\ \dots,\ x_{p-1}x_p,\ x_p^2 \right]$
- $\text{This takes up } N= (p + 1)^2 \text{ Which is a bit excessive and has duplication of terms } x_ix_j = x_jx_i \text{ otherwise the number of unique terms } 1 + p + \frac{p(p+1)}{2}$

### Example 2: The radial base function (RBF)

$$ K(u, v) = exp \{ - \frac{\mid \mid u - v \mid \mid ^2 _2 } {2\sigma^2}\}$$

- $\text{The mapping function } \varphi \text{ that affords this kernel can be derived and it maps instances to an inner product space of infinite dimension, } \\ \text{but we use this kernel without having to consider this mapping}$

- $\text{This function measure, similarity between data points such that:}$
    1. $\text{Close points obtain values near 1 } \mid \mid x - z \mid \mid^2 \approx 0 \rightarrow K(x,z) \approx 1$
    2. $\text{Distant points obtain values near 0 } \mid \mid x - z \mid \mid ^2 \approx 1 \rightarrow K(x, z) \approx 0$

### $\text{The role of Parameters}$

#### $\sigma - \text{Kernel width}$

1. $\text{Small - } \sigma \text{ OR Large - } \gamma$
    - $\text{Each training point has a very localized influence (The "bump" around each point is narrow)}$
    - $\text{The descision boundary becomes very wiggly - can perfectly fit training dat but RISKS overfitting}$
2. $\text{Large - } \sigma \text{ OR Small - } \gamma$ 
    - $\text{Each training point has a broad influence.}$
    - $\text{The descision boundary becomes smoother and more global}$
    - $\text{Risk of underfitting if too large - it may not capture fine distrinctions}$

##### $\text{Regularization Parameter (C, in SVM)}$

- $\text{Controls how strictly the classifier tries to seperate data vs. allow some miscalssification}$
- $\text{Together with } \sigma \text{it balanaces the bias-vairance tradeoff}$
    - $\text{Small } \sigma \ + \text{ Large C - Very tight fit (low bias and high vairiance)}$
    - $\text{Large } \sigma \ + \text{ Small C - Smoother descisionboundary (High bias, low vairance)}$

$\bold{Mercer's \ theorem } \text{ Provides a condition for establishing that K: } \mathbb{R}^p \times \mathbb{R}^p  \rightarrow \mathbb{R} \text{ is a valid kernel without having to specify mapping functions that affords it}$

1. **Symmetry condition**  
$$
K(x, z) = K(z, x) \quad \forall x, z
$$

2. **Positive semi-definiteness (PSD) condition**  
For any finite set of points $(\{x_1, x_2, \dots, x_n\})$ and any real coefficients $(\{c_1, c_2, \dots, c_n\})$:  
$$
\sum_{i=1}^n \sum_{j=1}^n c_i c_j K(x_i, x_j) \;\geq\; 0
$$

$\text{Below will show one the direct way (by definition) to prove a valid kernel}$
- $\text{To show that by the kernel function we can obtain our inner product mapping}$

#### Claim 
 $\text{The function K:} \mathbb{R}^p \times \mathbb{R}^p \rightarrow \mathbb{R} \quad K(u,v) = (1 + u^{\top}v)^2 \text{ is a kernel for this mapping}$

#### Proof: The Kernel Trick with a Quadratic Polynomial Kernel
$
\text{Let }  u, v \in \mathbb{R}^p  \\ \text{Define their feature-mapped versions as:} \\ \tilde{u} := \varphi(u), \quad \tilde{v} := \varphi(v) \in \mathbb{R}^N \\ \text{Then:} \\ \varphi(u)^\top \varphi(v) = \tilde{u}^\top \tilde{v} = \sum_{i=1}^N \tilde{u}_i \tilde{v}_i = \sum_{j=0}^p \sum_{l=0}^p u_j u_l v_j v_l \\ \text{On the other hand, consider the quadratic kernel:} \\ K(u, v) = (1 + u^\top v)^2 \\ \text{Now compute:} \\ \begin{aligned}
K(u, v) 
&= (1 + \sum_{j=1}^p u_j v_j)^2 \\
&= \left( \sum_{j=0}^p u_j v_j \right) \left( \sum_{l=0}^p u_l v_l \right) \quad \text{(with } u_0 = v_0 = 1 \text{ for bias)} \\
&= \sum_{j=0}^p \sum_{l=0}^p u_j u_l v_j v_l \\
&= \varphi(u)^\top \varphi(v)
\end{aligned}
$

### Example p = 2 : $x = (x_1, x_2)$

$\text{Applied by definition of the mapping that affords the kernel}$

$ \varphi (u) = \tilde u = (1, u_1, u_2, u_1, u_1^2, u_1u_2, u_2, u_2u_1, u_2^2) \in \mathbb{R}  \\
\varphi(u)^{\top} \varphi(v) = 1 + u_1v_1 + u_2v_2 + u_1v_1 u_1^2v_1^2 + u_1u_2v_1v_2 + u_2v_2 + u_1u_2v_1v_2 + u_2^2v_2^2 \\ \varphi(u)^{\top} \varphi(v) = 1 + u_1^2v_1^2 + 2u_1v_1 + 2u_2v_2 + 2u_1u_2v_1v_2 $

$ \text{We now use the kernel Trick}$

$ (1 + u^{\top}v)^2 = (1 + u_1v_1 + u_2v_2)^2 \\ = 1 + u_1^2v_1^2 + u_2^2v_2^2 + 2u_1v_1 + 2u_2v_2 + 2u_1u_2v_1v_2 \\ = \varphi(u)^{\top} \varphi(v)$

### Polynomial Kernels

- $\text{We can generalise the Kernel trick for the } \bold{full \ Rational \ variety} \text{ as follows:}$ $$ \boxed{K(u, v) = (1 + u^{\top}v)^r}$$

- $\text{We can change the constant 1, to } c \gt 0 \text{ producing a non-homgenous polynomial kernel}$ 

- $\text{If c = 0 then the Kernel becomes homogenous}$ $$ \boxed{K(u, v) = (c + u^{\top}v)^r}$$

### Dual Perceptron With Kernel 

1. $\text{Select } \bold{Kernel \ Function \ K:} \mathbb{R}^p \times \mathbb{R}^p \rightarrow \mathbb{R} \text{ based on the type of descision boundaries we wish to find}$
    - $\text{Example: } K(u, v) = (c + u^{\top}v)^r \text{ where our descision boundary is polynomial}$
2. $\text{Execute the } \bold{dual \ perceptron } \text{ algorithm with kernel K:} \\ \quad \bold{Initialise \ a = 0} \in \mathbb{R}^n \\ \quad \bold{Iterate} \text{ until termination} \\ \qquad \text{For i = 1, ..., n:} \\ \qquad \quad \bold{Predict:} \ \hat{y}^{(i)} = sign \left(\sum^n_{j=1}a_jy^{(j)} K(x^{(j)}, x^{(i)}) \right)\\ \qquad \quad \bold{Update: } \text{if } (y^{(i)} \ne \hat y^{(i)}) \text{ then } a_j = a_i + 2\eta $

- $\text{Complexity of algorithm depends on the complexity of the kernel function and the size of the support: } \mid \{i \mid a_i > 0 \}$
