# Multivariate Bayesian Classification

$\text{in the mult-feature setting, the class conditional } f_{X \mid Y} \text{ are multidimensional distribution } \mathbb{R^p}$

$\text{A multivariate continuous distributino over } \mathbb{R^p} \text{ is defined using a multidimension PDF } f_{X_1, \dots X_p}: \mathbb{R^p} \rightarrow \mathbb{R} \text{ which is non-negatie and intergrates to 1}$

- $\text{Class priors are still described using a single proabability vector } \pi$
- $\text{Class conditionals are described using p-dimensional PDFs } \{f_{X \mid Y = l}(x)\}^k_{l=1}$
- $\text{After fitting } \pi \text{ and } \{f_{X \mid Y = l}(x)\}^k_{l=1} \text{ to training data, classs prediction is done using the same approach as in the single feature case: }$
    - $c(x) = \argmin_j \left\{ \sum^k_{l=1} \pi_l f_{X \mid Y = l}(x) \lambda_{j, l}\right\}$

## Biggest Challange

$\text{How do we fit a p-dimensional class conditionals } \{f_{X \mid Y = l}(x)\}^k_{l=1} \text{ to the training data}$

$\text{A p-vairate normal distribution is defined by:}$ 
1. $\text{A mean vector } \mu \in \mathbb{R^p}$
2. $\text{A non-singular (invertible) covariance matrix } \sum \in \mathbb{R} \times \mathbb{R}$

$$ X \sim N(\mu, \Sigma) \iff f_X(x) = \frac{1}{\sqrt{(2\pi)^p \mid \Sigma \mid}}\exp \left\{-\frac{1}{2}(x - \mu) \cdot \Sigma^{-1} \cdot (x -\mu)\right\}$$

### How does this match the univariate case

- $\mu \text{ and } \Sigma \in \mathbb{R}$
- $\mid \Sigma \mid = \sigma^2 \text{ and } \Sigma^{-1} = \frac{1}{\sigma^2}$

$\text{This produces the first known guassian Curve}$

### What occurs when there 0 covariance between features

- $\text{We produce a diagonal matrix on } \Sigma$
- $\text{Each non-zero entry on the diagonal matrix corresponds to the diagonal to the variance of a perticular feature}$
- $ \mid \Sigma \mid = \prod^p_{t = 1} \sigma^2_t $
- $\frac{1}{2}(x - \mu) \cdot \Sigma^{-1} \cdot (x - \mu) = \sum^p_{t=1} \frac{(x_t - \mu_t)^2}{2\sigma^2_t}$

$$ \frac{1}{\sqrt{(2\pi)^p \mid \Sigma \mid}}\exp \left\{-\frac{1}{2}(x - \mu) \cdot \Sigma^{-1} \cdot (x -\mu)\right\} = \prod ^p_{t = 1} \frac{1}{\sqrt{2\pi \sigma^2_t}}\exp \left\{-\frac{(x_t - \mu_t)^2}{2\sigma^2_t}\right\}$$

## Naïve Bayes Approach

$\text{The conditionals } f_{X \mid y} \text{ are p-dimensional distributions }$
$\text{Fitting a p-dimensional distribution to data might be complex and can lead to overfitting } O(p^2) \text{ parameters}$

$\text{An alternative approach is to approaicamte each } P_{X \mid Y = y} \text{ using the product of p- one-dimensional distributions } O(p) \text{ parameters}$

$$ P_{X \mid Y = y}(x) = \prod^p_{t = 1} P_{X_t \mid Y = y}(X_t)$$


### Scenarios in which Naïve Bayes Helps

1. $\text{Suppose a student’s physics and chemistry grades are correlated: }$
$$
P(X_{\text{physics}}, X_{\text{chem}}) \neq P(X_{\text{physics}})P(X_{\text{chem}}).
$$

$\text{However, once we know the student’s math grade, the dependence may vanish:}$ 
$$
P(X_{\text{physics}}, X_{\text{chem}} \mid X_{\text{math}})
= P(X_{\text{physics}} \mid X_{\text{math}})\, P(X_{\text{chem}} \mid X_{\text{math}}).
$$

$\text{This is called}$  **conditional independence**.  

$\text{Naive Bayes relies on the assumption that features are conditionally independent given the class}$ (Y):  
$$
P(X_1, X_2, \dots, X_d \mid Y) \approx \prod_{i=1}^d P(X_i \mid Y).
$$

This simplification:  
- avoids estimating the full joint distribution,  
- reduces data requirements, and  
- makes classification efficient, even if exact independence does not strictly hold.
