## Single Feature (Univariate) MLE

$\text{We can fit class conditional PDFs } P_{X \mid Y} = f_{X \mid Y} \text{ seperately for each class either:}$
1. $\text{Non-parametric estimation G-KDE}$
2. $\text{Parametric Estimartion, using Guassians}$

$\text{ASSUMING, we have one of the two formulas defined we'll done it as follows } \{f_{X \mid Y = j} \}^k_{j=1}$
    - $\text{This is describing the probability of obtaining a perticular class (flower type), given that one feature value}$
$\text{To predict}$ $$ \argmax_j\{f_{X \mid Y = j} (x)\}$$

$\text{This doesn't take into consideration the prior information on the general population of the plant types themselves...}$

### Prior Probability 

$\text{This is the general probability on the classes, i.e. the ratio of the population on each plant type}$

$$ P(y = \text{Iris type}) = \frac{\# Iris Type Picked}{Sample \ Size} $$

### (Maximum) - Posterior Probabilities

$\text{This combines the Conditional probability with the Prior probability and normalises between [0,1], between all classes}$

$$ P_i = P( \text{Petal Size} > X \mid Y = \text{Iris Type}) \cdot P(Y = \text{Iris Type})$$

$$ \text{Posterior Probability } PP_i = \frac{P_i}{P_1 + P_2 + P_3}$$

$$ \text{MAP } = \argmax_j\{PP_j\}


## Training Bayesian Classifier

$\text{Training data } D = \{(x_i, y_i)\}^n_{i=1} \text{ can be used to fit class priors } (\pi_y) \text{ and class conditionals } (f_{X \mid Y}) \text{ that together define the joint probability of features(s) and the class label: }$ $$ P_{X,Y}(x,y) = \pi_y f_{X \mid Y}(x)$$

1. $\text{The MLEs of these parameters are obtained using the class proportions in the training data: (This is similar proof to the Binomial MLE)}$

$$ \hat \pi_j = \frac{1}{n}\sum^n_{i=1}1_{y_i = j}$$

2. $\text{The class conditionals are represented by k-PDFs:} $

$$ \{f_{X | Y=j}\}^k_{j=1}$$

$\text{Each PDF, is fit seperately to the subset of the training data correcponding to the class j} \\ \text{and this is done using either nonparametrix/paramertic estimation as presented in model estimation}$

$$ D_j = \{(x,y) \in D \mid y = j\}$$

## Classification Loss and Minimizing Risk 

$\text{The standard classification Loss function defined using class identity}$ $$L(y, \hat y) = 1_{\hat y \ne y}$$

$\text{So the risk of classifier c for data distributed according to } P_{X, Y} \text{ is the expected number of misclassification, similar to the MLE process we expand on the definition}$ 
$$ R(c; P_{X,Y}) = \mathbb{E_{P_{X,Y}}}[1_{c(x) \ne y}] = P(c(X) \ne Y) = \int\left(\sum^k_{l=1} \pi_l f_{X \mid Y = l}(x)1_{c(x) \ne l} \right)$$

$\text{To say we misclassified, is to say for a perticular class j, we obtain all other possible classification but j:}$ 

$$ \sum_{l \ne j}\pi_l f_{X \mid Y = l}(x) = \begin{align} \sum^k_{l=1} \pi_l f_{X \mid Y = l}(x) - \pi_j f_{X \mid Y = j}(x) \\ \sum^k_{l=1}P_{X,Y}(x,l) - \pi_j f_{X \mid Y = j}(x) \\ = f_X(x) - \pi_jf_{X \mid Y = j}(x)\end{align}$$

$\text{This is useful to know, since it means, we don't need to worry about if we're maximizing our function to minimize the risk,} \\  \text{the Bayseian Classifer as we know MAP, A-posterior indeed Minimizes the risk}$

$$ c(x) = \argmax_j \{\pi_j f_{X \mid Y = j(x)}\}$$

### Classification Loss Matrix

$\text{We can add further hyperparmeter costs, associated with misclassifications}$ 

$\text{Consider the case, where we have a general subset of some population being ill and healthy, in which we wish to determine if someone has a virus,} \\ \text{it would make sense that we wouldn't want to arrive at a case where someone is infected but we diagnosed them as healthy}$

$\text{To make sure this occurs, we add a } \lambda \text{ hyperparameter, with different weights such that, for mis-diagnosing we multiply the loss function by the respective weight of the } \lambda$

$\text{We create a matrix where it would be multiplied with the Posterior vector, along the diagnal will have a cost of 0, since we predicted correctly,} \\ \text{and for every type of misclassification we multiply by the repsective weight, } \\ \text{in the end we produce another vector in which we want to return the argument (class) of the minimum values}$

#### Classification Loss Matrix Example 

Consider the following loss Matrix: 

$$ 
\begin{bmatrix}
0 & 2 & 3 \\
1 & 0 & 3 \\
1 & 1 & 0
\end{bmatrix}


\begin{array}{c|ccc}
 & \text{True}_1 & \text{True}_2 & \text{True}_3 \\
\hline
\text{Pred}_1 & 0 & 2 & 3 \\
\text{Pred}_2 & 1 & 0 & 3 \\
\text{Pred}_3 & 1 & 1 & 0 \\
\end{array}
$$

$\text{Consider the following data: x = 5.1}$

$\text{Priors:} \quad \pi_{virg} = 0.15 \quad \pi_{vers} = 0.25 \quad \pi_{seto} = 0.6 $

$\text{Conditional probability:}  \quad f_{X \mid Y = virg} = 0.001 \quad f_{X \mid Y = virg} = 0.35 \quad f_{X \mid Y = virg} = 0.15 $

$\text{Posterior Probaility:} \quad P_{X,Y}(X, ) < 0.00015 \quad P_{X,Y}(X, ) = 0.088\quad P_{X,Y}(X,seto ) = 0.090 $

$\text{Currently at this point, we'd choose Seto, due to it's highest probability}$ 
- $\text{Note we didn't compute MAP yet, since it we'd need to sum and divide these values by the sum of the total values}$


$\text{But maybe, by misclassifying would cost us more so...}$

$$
c(x) = \arg\min_j \left\{ 
\begin{bmatrix}
0 & 2 & 3 \\
1 & 0 & 3 \\
1 & 1 & 0
\end{bmatrix}
\begin{bmatrix}
0.0001 \\
0.088 \\
0.090
\end{bmatrix}
\right\}
= \arg\min_j \left\{
\begin{bmatrix}
0.446 \\
0.270 \\
0.089
\end{bmatrix}
\right\}
$$
