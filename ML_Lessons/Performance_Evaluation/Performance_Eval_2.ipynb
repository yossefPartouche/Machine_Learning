## Confusion Matrix
$\text{A confusion matrix looks as follows}$ 

| $\text{Predicted} \rightarrow \\ \text{True } \downarrow$ | $\text{Positive}$ | $\text{Negative}$ | 
|-----------------------------------------|-------------------|-------------------|
| $\text{Positive}$ | $\colorbox{green}{ TP}$| $\colorbox{Red}{ FN}$| 
| $\text{Negative}$ | $\colorbox{Orange}{ FP}$ | $\colorbox{green}{ TN}$|

<br>

| $\text{Predicted} \rightarrow \\ \text{True } \downarrow$ | $\text{Positive}$ | $\text{Negative}$ | 
|-----------------------------------------|-------------------|-------------------|
| $\text{Positive}$ | $\colorbox{green}{ 3}$| $\colorbox{Red}{ 3}$| 
| $\text{Negative}$ | $\colorbox{Orange}{ 2}$ | $\colorbox{green}{ 6}$|


$\text{This matrix collects up the count, of the relation between predicted and actual values}$

$\text{General Pattern:} \\ \text{The first Word (True, False) describes match/mismatch with real value, The second word (Positive, Negatvie) describes what was predicted}$

$\text{TP (True Positive) := Actual Matches Prediction}$

$\text{TN (False Negative) := Actual (Positive) mismatched prediction (Negative)}$

$\text{TN ((True Negative) := Actual Matches Prediction)}$

$\text{FP (False Positive) := Actual (Negative) mismatched prediction (Positive)}$

 ### Key Performance Indicators (KPIs)

 $\text{Accuracy := } \frac{TP + TN}{N}$

 $\text{As mentioned earlier, this doesn't necessarily provide a clear cut answer of a "Good" Classifier, consider the following Matricies, describing potential results of corona tests}$

 $\text{They all produce the same accuracy (OK.. except the last one), clearly when diagnosing the patients we also want to consider the the number of FN and FP}$

 <center>

- $\text{Recall (True Positive Rate - TPR) :=} \frac{TP}{TP + FN}$
    - $\text{Proportion of ACTUAL POSITIVE count that was MATHCED}$

- $\text{False Positive Rate ( FPR) := } \frac{FP}{FP + TN}$
    - $\text{Proportion of ACTUAL NEGATIVE count that was MISMATCHED}$

- $\text{Prescision (Positive Predictive Value (PPV)) := } \frac{TP}{TP + FP}$
    - $\text{Proportion of POSITIVELY PREDICTION count that was MATHCED }$

- $\text{F1 Score:= } 2 \cdot\frac{recall \cdot prescision}{recall + prescision}$

$\text{This is a 2D graph, y - TPR and x - FPR, where we adjust the threshold [0,1] regular intervals to produce this plot graph}$

$\text{We then produce a best fit curve connecting the points}$
