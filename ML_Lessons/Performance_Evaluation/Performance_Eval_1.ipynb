- $\text{For us this means looking at any prediction model we've created and understand what are its limitations} \\ \text{be able to measure those limitations (so that we know in the future) and understand how to improve the model and at what cost}$

## Emperical Risk

$\boxed{\color{dodgerBlue}\text{Emperical Risk := The Expected error, measured on a loss function, } L}$
- $\text{This refers to a predictor, so it's a "risk on some predictor"}$

#### How it works

- $\text{We will usually be given: A loss function to compute the absolute deviation between the predicted value and the actual value}$
- $\text{The Emperical Risk, is just the sum of the loss function being applied for data sample and then divided by the size of data-set}$

### Emperical Risk Minimzation

$\text{We naturally want to reduce this risk as much as possible that is the ERM-Principle}$
$\text{Finding the varying aspect of the predictor function to minimize the risk}$
$\text{However, this adjustment can be imagined as a testing faze, meaning applying this to the training data set} \\ \text{can create a overfitting predictor}$ 

$\color{dodgerblue}\text{Overfitting := A predictor model performs "too well" on the data set and when tested performs bad on the validating/testing set}$

$\text{An overfitting predictor isn't good since it won't generalise well,} \\ \text{our main goal is to be able to understand the data set well enough so that the predictor model can produce reliable outcomes on unseen data}$

$\text{There are two types of errors we need to distinguish to generalise our models better}$

1. $\text{Training Error := Measures the risk predictor on the trainning data}$
    - $\text{in other words apply the Risk function on seen data}$
2. $\text{Generalisation Error := Measures the tisk of a predictor on NEW data}$
    - $\text{in other words apply the Risk function on unseen data}$

## Train/Test Split 

- $\text{We want to be able accurately measure the generalisation error using only the available data provided}$
- $\text{To do this we'll take a proportion of the training data and use it only in the testing stage}$

1. $\text{Split the Available data in two sets:}$ 
    - $\text{Training Set - } D_{train}$
    - $\text{Testing Set - } D_{test}$
    - $\text{Where - } D_{train} \cup D_{test} = D$
    - $\text{Ensure that  - } D_{train} \cap D_{test} = \empty$
2. $\text{Train a predictor } \hat y \text{using } D_{train} \text{ apply the ERM principle when evaluating}$ $$R_{train} := R(\hat y; D_{train})$$
3. $\text{We then evaluate to obtain an approximation of the generalisation error: }$ $$R_{test} := R(\hat y; D_{test})$$

    1. $D_{validate}$

### But is one validation Enough? (Cross Validation)

- $\text{If our ultimate goal is to roduce a minimizing predictor, then this (ultimate) predictor is dependant on the number of Train/Validation Splits}$
- $\text{So if we can create more splits in different ways then we can actually create a more robust predictor}$

#### How it works: 
1. $\text{First split the test set aside}$
2. $\text{Divide the Trainning set into m disjoint (roughly) equal subsets}$
3.	$\text{For each fold } i \text{(where } i = 1, \dots, m):$
    - $\text{Use fold } i  \text{ as the validation set.}$
	- $\text{Use the remaining  } m-1  \text{ folds as the training set.}$
	- $\text{Train the model and compute the validation error (risk) on fold } i$
4. $\text{Compute the average validation error across all } m  \text{ folds. This is the cross-validation error.}$
5. $\text{Use this to choose the best model or hyperparameters.}$
6. $\text{Finally, evaluate the chosen model on the test set to estimate generalization error.}$
