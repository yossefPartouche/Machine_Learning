## Anomaly Detection in Network Traffic

$\text{Security analysts montior network traffic to identify unusual activity, such as distributed denial of service attack (DDoS)}$

$\text{A useful indicator suspicious acticity in the network packet rate (number of packets per second)}$

$\text{If we're able to determine and understand the typical packet rate of a legitimate user, then we can mitigate such attacks}$

### Non-Parametric Estimation 

1. $\text{We collect data from X (50) legitimate users}$
    - $\text{We assume that the data has a distribution with PDF } f_X$
2. $\text{We use the data to compute } \hat f \text{ an estimated version of } f_X$
3. $\text{Then use } \hat f \text{ as our model to detemine the probabilty that a user is legit or suspicsous}$
    - $P(X \geq x) = \int_{x}^{\infty} \hat{f}(t)dt $

### Parametric Estimation 

$\text{Assume our data fits a parametric from for the true and unknown distribution (single class)} f_{X \mid \theta}$

1. $\text{Find an estimation } \hat \theta \text{ on the parameters } \theta \text{ and use } \hat f = f_{X \mid \hat \theta}$

$\text{Suppose we have the following data of 7 packet rates from legitimate users: } D = \{5.3,\ 4.1,\ 10.2,\ 4.3,\ 2.5,\ 5.9,\ 6.2,\}$

$\text{Assuming these rates are independently samples froma normal distrbuition } N(\mu, V), \text{ we compute the MLE for } \hat \theta = (\mu, \sigma^2)$

1. $\hat \mu = \bar x = \frac{1}{7}\sum^7_{i=1} x_i$
2. $\hat \sigma^2 = \frac{1}{7} \sum^7_{i=1} (x_i - \hat \mu)^2$

$\text{We now have the parameters to create Bell curve, and thus predicting}$

### Parametric Estimation Multiple Classes

$\text{Training: }$
- $\text{For each class c, we fit a seperate Guassian } \hat \theta_c = (\hat \mu_c, \hat \sigma^2_c)$


$\text{Prediction: }$
- $\text{given A new instance x, look for } H_{\hat \theta_c} \text{ that maximizes the posterior probabiltiy assuming all classes }$

If you want the class that maximizes the class-conditional Gaussian likelihood (using estimated parameters \hat\mu_c,\hat\sigma_c^2), the corrected formula is

$$
\hat y(x)= \arg\max_{c}{f_{X\mid H_{\hat\theta_c}}(x)}
= \arg\max_{c} \frac{1}{\sqrt{2\pi,\hat\sigma_c^{2}}},
\exp\left(-\frac{(x-\hat\mu_c)^2}{2\hat\sigma_c^{2}}\right).
$$

Two useful variants you’ll often see:

1. MAP (include class priors P(c)) — choose class that maximizes posterior:
$$
\hat y(x)=\arg\max_c; P(c),f_{X\mid c}(x)
=\arg\max_c; P(c),\frac{1}{\sqrt{2\pi,\hat\sigma_c^{2}}}
\exp!\left(-\frac{(x-\hat\mu_c)^2}{2\hat\sigma_c^{2}}\right).
$$

2. Numerically stable (log form) — take logs to avoid tiny numbers and simplify:
$$
\hat y(x)=\arg\max_c;\Big[
\log P(c) -\tfrac{1}{2}\log(2\pi\hat\sigma_c^{2})
-\frac{(x-\hat\mu_c)^2}{2\hat\sigma_c^{2}}
\Big],
$$
and for the pure likelihood (no priors) drop the \log P(c) term.

## Click through rate of online ads

$\text{The CTR is the prbability a user click on an ad when it pops up}$

$\text{We wish to estimate the CTR of an ad from data, so that we could:}$
- $\text{Compare differrent ads and see which performs better}$
- $\text{Allocate budget to higher performing ads}$
- $\text{Predict future revenue based on expected clicks}$

$\text{An ad with CTR of p that was displayed n times (to different users) will be clicked X times where } X \sim Binom(n, p)$

$\text{This distribution } P_X \text{ is defined by a single parameter } \theta = p \in [0, 1]$

$\text{We wish to determine the value of } \theta = 0 \text{ which maximizes the likelihood } P_{X \mid \theta} (x)$

---

### Likelihood

$\text{The setting:}$
- $\text{A probabilitic model } H_\theta \text{ specified using a parameter } \theta \in \R^d$
- $\text{Data D = } \{x_i\}^n_{i=1} \text{ assumed to be generated by } H_{\theta}$

$\text{The liklihood function } L: \mathbb{R}^d \rightarrow [0,1] \text{ is defined as:}$
- $L(\theta, D) = P_X(D \mid H_{\theta})$
- $\text{Where } L(\theta, D) \text{ is the probabiliy of observing D under the model } H_{\theta}$

#### Relating to CTR Example 

- $H_{\theta} = X \sim Binon(n,p) \mid \ \theta = p $
- $\text{If we observe D} = \{x\} \text{ clicks in n displays the liklihood is: } L(\theta; D) = P(X = x \mid H_{\theta}) = \binom{n}{x} \theta^x(1-\theta)^{n-x}$

### Maximizing Liklihood 

$\text{Indeed since the definiton of the likelihood describes the probability of observing the data under a perticular model, we wish to maximiza this proability}$
$\text{This is known as the Maximum Likelihood Estimator (MLE), which determines the maximum value of } \theta$
$$ \theta^* = \argmax_\theta L(\theta, D)$$

$\text{This can be done using gradient descent or Anayltically}$

---

### MLE for Binomial Distrbution

$ \theta ^* = \argmax_{\theta}L(\theta, D) = \argmax_{\theta} \binom{n}{x} \theta^x (1-\theta)^{n-x}$

$$
L'(\theta) 
= \binom{n}{x} \Big( x \theta^{x-1} (1-\theta)^{n-x} - \theta^x (n-x)(1-\theta)^{\,n-x-1} \Big) 
$$

$$
= \binom{n}{x} \, \theta^{x-1} (1-\theta)^{\,n-x-1} \, \Big( x(1-\theta) - \theta(n-x) \Big)
$$

$$
= \binom{n}{x} \, \theta^{x-1} (1-\theta)^{\,n-x-1} \, \Big( x - \theta x - \theta n + \theta x \Big)
$$

$$
= \binom{n}{x} \, \theta^{x-1} (1-\theta)^{\,n-x-1} \, (x - \theta n)
$$

$\text{Our potential solutions are 0, 1 and } \frac{x}{n}$

$\text{When substituting each of these values back into L, we see that } \frac{x}{n} \text{ produces that highest results and is the maximizer}$

## Click through rate ad online - 2

$\text{Now we examine the number of click on an add in two seperate time intervals:}$

$\text{Our model: } H_p : X_i \sim Binom(n_i, p) \ i = 1,2$

$\text{Data: D = } \{(n_1, x_1), (n_2, x_2)\} \text{ assume independence}$

$L(\theta; D) = P(X_1 = x_1, X_2 = x_2 \mid H_p) = P(X_1 = x_1 \mid H_p)P(X_2 = x_2 \mid H_p) = \binom{n_1}{x_1} \ \theta^{x_1}(1-\theta)^{n_1-x_1} \binom{n_2}{x_2} \theta^{x_2}(1-\theta)^{n_2-x_2}$

$\text{By additivity of joint binomial we have that MLE } \hat \theta = \frac{x_1 + x_2}{n_1 + n_2}$

### Generalised to many independent sampling

- $ D = \{x_i\}^n_{i=1} \text{ where each } x_i \text{ is independently sampled from } P_\theta$
- $ \text{The likelihood in such a case } L(\theta; D) = P(X_1 = x_1, X_2 = x_2, \dots, X_n = x_n \mid H_\theta) = P_\theta(x_1)P_\theta(x_2) \dots P_\theta(x_n) = \Pi^n_{i=1}P_\theta(x_i)$
- $\text{since the log function is a great way to convert between the products to the sum of terms, whilst maintainting the integrity of the data we have the Log-Likelihood}$
$$ l(\theta, D) := \log[(\theta; D)] = \sum^n_{i=1} = log[P_{\theta}(x_i)]$$

## Confidence in statistical Estimation

$\text{After Computing an estimated } \hat \theta \text{ we would like to know how far it is from the true parameter values } \theta$
- $\text{A common way to quantify confidence in an estimate is to specify a } Confindence \ Interval \text{ CI, that contains the true } \theta \text{ with some true pre-specfied high probability }\\ 1 - \alpha \text{ (0.9, 0.95, 0.99)} $
- $\text{This is done by characterising the distribution of the estimated value } \hat \theta \text{ in terms of } \theta$

### CI for Normal Mean Estimation 

- $\text{Assume each data instance in } D = \{x_1, x_2, \dots, x_n \} \text{ is sampled independently from the same normal distrubtion } N(\mu, \sigma^2)$
- $\text{The MLE of the normal mean is } \hat \mu = \bar x = \frac{1}{n}\sum^n_{i=1}x_i \text{ is obtained by averaging n i.i.d. normal variables } X_i \sim N(\mu, \sigma^2)$
    1. $\text{We now consider the following properties on normal variable if }$ $$ X \sim N(\mu, \sigma^2) \text{ and } a \text{ scalar constant then }  aX \sim N(a\mu, a^2\sigma^2)$$ 
    2. $\text{If }$ $$ X_1 \sim N(\mu_1, \sigma^2_1) \text{ and } X_2 \sim N(\mu_2, \sigma^2_2) \text{ and are independent then } X_1 + X_2 \sim N(\mu_1 + \mu_2, \sigma^2_1 + \sigma^2_2)$$

    3. $\text{This implies that...}$ $$ \hat \mu = \frac{1}{n} \sum^n_{i=1} X_i \sim N(\frac{n}{n}\mu, \frac{n}{n^2}\sigma^2) = N(\mu, \frac{1}{n}\sigma)$$

- $\text{The MLE of } \mu \text{ is } \hat \mu = \frac{1}{n}\sum^n_{i=1}x_i$
- $\text{The MLE of } \sigma^2 \text{ is } \hat \sigma^2 = \frac{1}{n}\sum^n_{i=1}(x_i-\hat \mu)^2$

$\text{When n is sufficiently large we have that:}$

$$\frac{\sqrt(n) ( \hat \mu - \mu )}{\hat \sigma} \sim N(0,1)$$

$\text{From this we can determine the Confidence interval, for a given confidence level } (1- \alpha) \text{ compute } z = \phi^-1(1 -\frac{\alpha}{2}) P(\frac{\sqrt n \mid \hat \mu - \mu \mid}{\hat \sigma} \ge z) = 2(1 - \phi(z)) =  2(1 - (1 - \frac{\alpha}{2})) = \alpha$

### CI Normal Mean Estimation Example 

$D = \{5.3, \ 4.1, \ 10.2,\ 4.3,\ 2.5\, 5.9,\ 6.2,\}$

$\text{Assume these rates are i.d.d. sampled from } N(\mu, \sigma^2)$

1. $ \hat \mu = \bar x = \frac{1}{7}(5.3 + 4.1+ 10.2 + 4.3 + 2.5 + 5.9 + 6.2 ) = \boxed{5.5}$

2. $\hat \sigma = \left( (5.5 - 5.3)^2 + (5.5 - 4.1)^2 + (5.5 - 10.2)^2 + (5.5 - 4.3)^2 + (5.5 - 2.5)^2 + (5.5 - 5.9)^2 + (5.5 - 6.2)^2 \right)^{\frac{1}{2}} = \boxed{2.24}$

3. $\text{Compute z for confidence level } 1-\alpha = 95 \%$ $$ z =\phi^{-1}(1 - \frac{\alpha}{2}) = \phi^{-1}(0.975) = 1.96$$

4. $\text{The 95\% CI for } \mu \text{ is } [\hat \mu - \frac{\hat \sigma}{\sqrt(n)}z, \hat \mu + \frac{\hat \sigma}{\sqrt(n)}z] $

$$[5.5 - \frac{2.24}{\sqrt(7)}(1.96), 5.5 + \frac{2.24}{\sqrt(7)}(1.96)] = [3.84, 7.16]$$

## Comparing two Classifiers

$\text{The generalisation error of classifier x is defined as: } p := p(y \ne c(X)) \\ \text{So, the empirical risk provides the MLE for the generalisation error p } \hat p = R(c; D_{test}) = \frac{n_{err}}{n}$

$\text{Our Null Hypothesis } H_0 \text{ if both classifiers have the same true error rate } R_A = R_B$

$\text{We then compute the difference in observed risks, we can then transform to follow Normal Distribution and then answer the following question:}$

$$\text{If the Resulting z-score if large (absolute), the probability of seeing that difference under our hypothesis } H_0 \text{ is very small, so we reject the hypothesis}$$

$\text{We'll now present the method with real examples}$

Case A: Classifiers evaluated over a test set with 100 samples: 

| | Classifier 1 | Classifier 2| 
|-|---------------|------------|
| # errors | 11 | 15 |
| Risk $R(C; D_{test})$ | 0.11 | 0.15|

Case B: Classifiers evaluated over a test set with 1000 samples:

| | Classifier 1 | Classifier 2| 
|-|---------------|------------|
| # errors | 110 | 150 |
| Risk $R(C; D_{test})$ | 0.11 | 0.15|

$\text{We wish to assess whether the two classifers could have identival generalisation error}$

$\text{Assuming two classifiers } c_1 \ c_2 \text{ with GENERALISATION ERRORS } p_1 \ p_2 \text{ we wish to determine the better of the two, since we DON'T know } p_1, p_2 \text{ we estimate using } R(c_i, D_{test})$

$\text{We check using } \hat p_i \text{ that } \mid \hat p_1 - \hat p_2 \mid \text{ is large compared to what is plausibale when } p_1 = p_2$

$\text{Using the Central Limit theorem, given the samples are large enough n, we can convert both towards normal distribution } \sqrt(n)(\hat p_i - p) \sim N(\mu = 0, \sigma^2 = p(1-p))$

### Method

1. $\text{Compute emprical Risk on n test data set instances}$
$$ \hat p_1 =  R(c_1; D_{test}) = \frac{n_{err_1}}{n_1} \quad \hat p_2 =  R(c_2; D_{test}) = \frac{n_{err_2}}{n_2}$$

2. $\text{Compute the } \mathbb{standardized} \text{ difference}$ 
$$ z_{1,2} := \frac{\sqrt(n)(\hat p_1 - \hat p_2)}{\sqrt(2 \hat p(1 - \hat p))} \quad \hat p = \frac{n_{err_1} + n_{err_2}}{2n}$$

3. $\text{The probability to get a value larger than } \mid z_{1,2} \mid \text{ when } p_1 = p_2 \text{ is } 1 - 2\phi(\mid z_{1,2}\mid)$

    - $\text{if this value is small, it's unlikely that } p_1 = p_2 \text{Which means they're not easily compariable classifiers by this method}$

## Case A: 100 samples  

Two classifiers evaluated over a test set with **100 samples**:  

| Case A | Classifier 1 | Classifier 2 |
|--------|--------------|--------------|
| #errors | 11 | 15 |
| Risk $R(c; D_{test})$ | 0.11 | 0.15 |

$$
|Z_{1,2}| = \frac{\sqrt{n} \, |\hat{p}_1 - \hat{p}_2|}{\sqrt{2 \hat{p}(1-\hat{p})}} 
= \frac{\sqrt{100} \, |0.11 - 0.15|}{\sqrt{2 \cdot 0.13 (1-0.13)}} 
\approx 0.84
$$

$$
\mathbb{P}(|Z_{1,2}| > 0.84 \mid p_1 = p_2) 
\approx 2\Phi(-0.84) \approx 0.4

\\

\text{This means the observed difference is very fairly likeli under the null hypothesis were true...Suggests the classifiers are not so different}
$$
