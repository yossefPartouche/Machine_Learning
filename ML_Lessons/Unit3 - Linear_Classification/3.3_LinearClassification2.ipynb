## Logistic Regression (LoR)

- $\text{Notation switch instead of our binary classes being \{-1, +1\} it'll be changed to \{0, 1\}}$

### Probablistic Model for Linear Separation

$\text{Given w, each data point x is associated with a pair of probabilities:}$ $$p(y \mid x, w) = \left(Pr(y = 0 \mid x, y), Pr(y = 1 \mid x, w) \right ) \mid \left ( \sum^1_{i=0}Pr(y = i \mid x , w) = 1 \right)$$

$\text{We'll make the following assumptions:}$ 
- $\text{For } w^{\top}x = 0 \text{ we have } p(y \mid x, w) = (0.5,0.5)$
- $\text{For } w^{\top}x \rightarrow ∞ \text{ we have } p(y \mid x, w) = (0,1)$
- $\text{For } w^{\top}x \rightarrow -∞ \text{ we have } p(y \mid x, w) = (1,0)$
- $ w^{\top}x_1 = w^{\top}x_2 \rightarrow p(y \mid x_1, w) = p(y \mid x_2, w) \\ \text{ where } p(y = 1 \mid \vec{x}, w) \text{ is a monotonous increasing function } w^{\top}\vec{x}, \text{ which from now we'll denote by } \sigma(w^{\top}\vec{x})$
    - $\text{In essence this means that we'll be looking at the model as a probablistic function of y = 1,} \\ \text{which is conditioned (dependant) on the features, x and the parameters, w}$

#### Desired Properties of $\sigma(t)$
1. $\sigma(t) + \sigma(-t) = 1$
2. $\lim_{t \rightarrow ∞}\sigma(t) = 1$
3. $\text{A desired shape of the function:}$

### The Sigmoid Function

$\text{We'll present two equivelant definitions of the same function}$ 

1. $ p(y = 1 | x, w) = \sigma(w^{\top}x)$
2. $  \sigma(t)= \large \frac{1}{1 + e^{-t}}$

- $\text{So when we say } \sigma(t) \text{ we're short-noting} \\  \text{"The probability of classiying our sample point to be 1 given the following features x and the set parameters w"}$

#### Confirm the Desired Properties are met 
1. $$\sigma(t) + \sigma(-t) = \frac{1}{1 + e^{-t}} + \frac{1}{1 + e^{t}} = 1$$

2. $$\lim_{t \rightarrow \infty} \sigma(t) = \lim_{t \rightarrow \infty} \frac{1}{1 + e^{-t}} = \frac{1}{1 + 0} = 1$$

3.  **Derivative of the Sigmoid Function**  
$$
\begin{aligned}
\frac{d}{dt} \sigma(t) &= \frac{d}{dt} \left( \frac{1}{1 + e^{-t}} \right) \\
&= - (1 + e^{-t})^{-2} \cdot \frac{d}{dt}(e^{-t}) \\
&= (1 + e^{-t})^{-2} \cdot e^{-t} \\
&= \frac{1}{1 + e^{-t}} \cdot \frac{e^{-t}}{1 + e^{-t}} \\
&= \frac{1}{1 + e^{-t}} \cdot \left(1 - \frac{1}{1 + e^{-t}} \right) \\
&= \sigma(t) \cdot (1 - \sigma(t)) \text{ which is maximized when } \sigma(t) = 0.5
\end{aligned}
$$
4. $\text{Assume that } p(y = 1 \mid x, w) = \sigma(w^{\top}x) \mid \sigma(t) = \frac{1}{1+e^{-t}} \text{ then:}$
$$\begin{aligned}
\log\left(\frac{\sigma(t)}{1 - \sigma(t)}\right) = t \iff \\ \frac{\sigma(t)}{1 - \sigma(t)} = e^t \iff \\ \sigma(t) = e^t(1 - \sigma(t)) \iff \\ \sigma(t)(1 + e^t) = e^t \iff \\ \sigma(t) = \frac{e^t}{1+e^t}=\frac{1}{1+e^{-t}}\end{aligned}$$

$ \text{Conclusion:} \\ \boxed{\text{log-odds ratio :=  } \log \left ( \frac{p(y = 1 \mid x, w)}{p(y = 0 \mid x, w)} \right) = w^{\top}x}$



#### What does Logistic regression do?

- $\text{Logistic regression attempts to find the hyperplane w that provides the bet linear } \bold{approximation} \text{ of the log-odds ratio}$
- $\text{Using this probablistic model, we now want to find that parameters w, which will maximize the likelihood for a given dataset}$



## Optimization for Logistic Regression 

$ \text{Given D :=} \{(x^{(i)}, y^{(i)})\}_{i=1}^n \text{ where } x^{(i)} \in \mathbb{R}^{p+1}, y^{(i)} \in \{0, 1\} \text{ and the hyperplane } w \in  \mathbb{R}^{p+1} \text{, then the log-likelyhood is:}\\ \text{(Note that the bias parameter will have the 0-th feature be 1 as usual)}$

$$
\begin{aligned}
l(w; D) &= \sum_{i=1}^n\log[P_{XY \mid w}(x^{(i)}, y^{(i)})] \\
&= \sum_{i=1}^n\log[P_{X\mid w}(x^{(i)})P_{Y\mid X, w}(y^{(i)})] \\
&=\sum_{i=1}^n\log[P_{X\mid w}(x^{(i)})] + \sum_{i=1}^n\log[P_{Y\mid X, w}(y^{(i)})] \\
&= \sum_{i=1}^n\log[P_{X}(x^{(i)})] + \sum_{i=1}^n\log[p(y = y^{(i)} \mid x^{(i)}, w)] \\
&= \sum_{i=1}^n\log[P_{X}(x^{(i)})] + \sum_{i=1}^n\log[p(y = 1 \mid x^{(i)}, w)] + \sum_{i=1}^n\log[p(y = 0 \mid x^{(i)}, w)] \\
&= \sum_{i=1}^n\log[P_{X}(x^{(i)})] + \sum_{i=1}^n \left ( y^{(i)}\log[\sigma(w^{\top}x)] + (1 - y^{(i)})\log[1 - \sigma(w^{\top}x)] \right)
\end{aligned}
$$

$\text{Our Goal:}$
$$ \\ \large \argmax_{w \in \mathbb{R}^{p+1}}\sum_{i=1}^n \left ( y^{(i)}\log[\sigma(w^{\top}x)] + (1 - y^{(i)})\log[1 - \sigma(w^{\top}x)] \right)$$

$\text{Note that since } \sigma(w^{\top}x) \in [0,1] \text{ so the } \log \text{ will produce negative values as such it's easier to solve this issue as a minimization problem by:}$

$$ \large \argmin_{w \in \mathbb{R}^{p+1}}\sum_{i=1}^n \left ( - y^{(i)}\log[\sigma(w^{\top}x)] - (1 - y^{(i)})\log[1 - \sigma(w^{\top}x)] \right)$$


### Binary Cross Entropy 

$\text{The binary entropy function of a binary distribution } (p, 1-p):$ $$h(p) := -p\log(p) - (1-p)\log(1-p)$$

$\text{The binary cross-entropy (BCE) between two binary distributions } (p, 1-p), (q, 1-q):$ $$h(p,q) := -p\log(q) - (1-p)\log(1-q)$$

$\text{For us we have 2 distributions:}$
1. $\text{The label (class)}$
    - $p = 1 \rightarrow 1-p = 0$
    - $(p, 1-p) \rightarrow (1,0)$
2. $\text{The dependency between the features and the parameters:}$
    - $ q = p(y=1 \mid x^{(i)}, w) = \sigma(w^{\top}x^{(i)})$
    - $ 1 - q = 1 - p(y=1 \mid x^{(i)}, w) = 1- \sigma(w^{\top}x^{(i)})$

- $\text{To understand deriving the BCE minizes the BCE loss follow the }$ BCE\_Loss.ipynb $\text{ notebook}$

### Gradient Descent for LoR

$$ l(w; D) = \sum_{i=1}^n \left ( - y^{(i)}\log[\sigma(w^{\top}x)] - (1 - y^{(i)})\log[1 - \sigma(w^{\top}x)] \right) $$


| **Full-Batch BCE** | **Stochastic BCE (SGD)** | **Mini-Batch BCE** |
|--------------------|--------------------------|---------------------|
|$\begin{aligned}&\textbf{1. Initialize: } \mathbf{w}^{(0)} \in \mathbb{R}^{p+1},\ t = 0 \\ &\textbf{2. Compute prediction: } \hat{y}^{(i)} = \sigma((\mathbf{w}^{(t)})^\top \mathbf{x}^{(i)}) \\ &\textbf{3. Compute full gradient: } \nabla J(\mathbf{w}) = \frac{1}{n} \sum_{i=1}^{n} \left( \hat{y}^{(i)} - y^{(i)} \right) \mathbf{x}^{(i)} \\ &\textbf{4. Update weights: } \mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \eta \nabla J(\mathbf{w}^{(t)}) \\ &\textbf{5. Repeat until convergence.} \end{aligned}$ | $\begin{aligned} &\textbf{1. Initialize: } \mathbf{w}^{(0)} \in \mathbb{R}^{p+1},\ t = 0 \\ &\textbf{2. For each training sample } (x^{(i)}, y^{(i)}): \\ &\quad \bullet\ \hat{y}^{(i)} = \sigma((\mathbf{w}^{(t)})^\top \mathbf{x}^{(i)}) \\ &\quad \bullet\ \text{Gradient: } g^{(i)} = \left( \hat{y}^{(i)} - y^{(i)} \right) \mathbf{x}^{(i)} \\ &\quad \bullet\ \text{Update: } \mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \eta \cdot g^{(i)} \\ &\textbf{3. Repeat for next sample or next epoch.} \end{aligned}$ | $\begin{aligned} &\textbf{1. Initialize: } \mathbf{w} \in \mathbb{R}^{p+1} \text{ randomly} \\ &\textbf{2. ## Multi-Class Classification

$ \text{Given D :=} \{(x^{(i)}, y^{(i)})\}_{i=1}^n \text{ where } x^{(i)} \in \mathbb{R}^{p+1}, y^{(i)} \in \{0, 1, \cdots ,k\}$

$\text{To use logistic regression to best approximate the classification:}$
1. $\text{We train k-seperate binary classifiers creating k-hyperplanes } w^{(1)} \cdots w^{(k)}$
    - $\text{When we applied that the following proability: } p(y = 1 | \cdots) \\ \text{ We are saying "what is the probability that this is the class we're looking for or not."}$
    - $\text{So when we train k-classifiers we do the same thing but fo, LoR}$
2. $\text{We then create a } (p+1)$ x $ k \text{ matrix W}$ 
    - $\text{Each row corresponds to the hyperplane of a perticular classifier}$
    - $\text{So that } \sigma([W^{\top}x]_j) = p(y = j \mid x , w)$

3. $\text{Finally to predict the highest likely class we chose the Max index as the predicted label}$
