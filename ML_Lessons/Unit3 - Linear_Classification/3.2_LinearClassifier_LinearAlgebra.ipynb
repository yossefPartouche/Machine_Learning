# Geometric interpretation of Linear Classification

$\text{We will focus in } \R^2 \text{ linear discriminant function to understand the goemetric interpretation but as everything is vectorised it's applicable to any dimension}$

$\text{We will break down the basics to make things more visual and thus more understandable}$

$\begin{pmatrix}
w_0 \\
w_1 \\
w_2
\end{pmatrix} = \vec{w} \in \R^3 \\ \begin{pmatrix}
1 \\
x_1 \\
x_2
\end{pmatrix} = \vec{x} \in \R^3$

$f_w(x) = w^{\top}x = \begin{pmatrix}
w_0 &
w_1 &
w_2
\end{pmatrix} \cdot \begin{pmatrix}
1 \\
x_1 \\
x_2
\end{pmatrix} = w_0 \cdot 1+ w_1 \cdot x_1 + w_2 \cdot  x_2$

- $\vec{w} \in \R^3 \text{ are the parameters we're aiming to find with the algorithm and } \\  \vec{x} \in \R^3 \text{ is are feature vector which we substitute into our function to classify the sample point}$
- $\text{So if we assume that we obtained our } \vec{w} \in \R^3 \text{parameters, then given any sample point } \begin{pmatrix} x_1, & x_2 \end{pmatrix} \\ \text{ (notice we don't include the } x_0 \text{ bias term) } \\ \text{then any point on the line (hyperplane) will satisfy the equation } f_w(x)=0$

|$$\large w_0 + w_1x_1 + w_2x_2 = 0$$|  $$x_2 = \boxed{-\frac{w_1}{w_2}x_1 - \frac{w_0}{w_2}}$$|
|-|-|
$\text{On the LHS we have the implicit form of the equation} \\ \text{(describing whether our sample point is on the line)}$ | $\text{On the RHS we have the explicit form of the equation} \\ \text{(describing where one of the features of the sample points will be in relation to the line)} \\ \text{You can imagine this form as } y = mx + c$|

### Focus on the implicit form
$\text{The implicit Representation presents a more general idea and robust method for point classification}$

- $\text{We know that if the dot product between two vectors = 0 then they are orthogonal (perpendicular from another)}$
- $\text{So when we define the function } f_w(x) = w^{\top}x + w_0 \text{ we're defining a hyperplane (line in this case)}$
- $\text{To be specific } \vec{w} \text{ defines the orientation of the hyperplane and specifically } w_0 \text{ defines its position relative (shortest orthogonal distance) to the origin}$
- $ \text{So with the above in mind it leads to the following known equation: } \boxed{w^{\top}x = || \vec{w} || \cdot || \vec{x} ||\cdot \cos(\theta)}$
    - $\text{With a bit of manipulation we obtain the following: } \boxed{\frac{w^{\top}x}{|| \vec{w} || \cdot || \vec{x} ||} = \cos(\theta)} \text{ which leads to the following cases}$

#### How geometry of Perceptron Algorithm change our hyperplane

$\text{Reminder:}$

$
\begin{aligned}
&\text{1. Start with a random } \mathbf{w} \in \mathbb{R}^{p+1}. \\
&\text{2. Repeat until convergence (correctly classified):} \\
&\quad \text{2.1. For each training sample } (\mathbf{x}^{(i)}, y^{(i)}) \in \mathcal{D}: \\
&\qquad \text{2.2. If } \text{sign}(\mathbf{w}^\top \mathbf{x}^{(i)}) \neq y^{(i)}, \text{ adjust } \mathbf{w}: \\
&\qquad \mathbf{w} \leftarrow \mathbf{w} + \Delta \mathbf{w}, \text{ where } \\
&\qquad \color{dodgerblue}{\boxed{\Delta \mathbf{w} = -\eta \big( \text{sign}(\mathbf{w}^\top \mathbf{x}^{(i)}) - y^{(i)} \big) \mathbf{x}^{(i)}.}}
\end{aligned}
$



$\text{The final line in the algorithm is what will lead to the change of the hyperplane}$
- $\text{since } w \text{ is a vector which describes the direction of the hyperplane when we apply the } \Delta w \\ \text{ we're actually describing a change in direction of the hyperplane OPPOSITE to the direction from which there was a misclassification.}$
- $\text{Note if there wasn't a misclassification the } \Delta w = \vec{0} \text{ since sign}(\mathbf{w}^\top \mathbf{x}^{(i)}) - y^{(i)} = 0 \text{ and as such } \eta \text{ and } x^{(i)} \text{don't affect it, which is how we desire it}$
