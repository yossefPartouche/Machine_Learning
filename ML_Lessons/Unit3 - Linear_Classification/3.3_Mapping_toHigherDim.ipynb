# Mapping to Higher Dimension

$\text{What about non-Linear classification; i.e. creating a non-linear descision boundary?}$

$\text{General idea:}$

- $\text{We check if in the given dimension of the number of features, we can obtain linear seperability}$
- $\text{If it's not linearly sepeable then we create a mapping which will increase the dimension of the feature space}$
    - $\phi : \R^p \rightarrow (\R^p, \_ , \_ , \cdots ,\_)$
- $\text{We then check again (over this dimension) for a linear sperable hyperplane}$

### Common Mapping examples

| Feature Dimension | Class Dimension | Mapping Function | 
| -| - | - | 
| $ x \in \R$ | $y \in \R $ | $$ \varphi(x) = (x, x^2)$$ | 
| $ x \in \R^2$ | $y \in \R $ | $$ \varphi(x) = (1, x_1, x_2, x_1x_2)$$ | 
| $ x \in \R^2$ | $y \in \R $ | $$ \varphi(x) = (1, x_1, x_2, x_1^2, x_2^2)$$ | 
| $ x \in \R^2$ | $y \in \R $ | $$ \varphi(x) = (1, x_1x_2) $$ | 
| $ x \in \R^2$ | $y \in \R $ | $$ \varphi(x) = (1, x_1^2, x_2^2)$$|
| $ x \in \R^2$ | $y \in \R $ | $$ \varphi(x) = (1, x_2, x_1^2x_2)$$ |

### Rational Varieties

$\text{Above we've noticed different methods to map to a higher dimension and we've done so up to a certain degree(r)}$
$\text{This is known known as the full rational variety known examples are:}$
- $\text{Quadratic Variety } x \in R^2: \quad \varphi{(x_1, x_2)} = (1, \ x_1, \ x_2, \ x_1^2, \ x_2^2, \ x_1x_2)$
- $\text{Cubic Variety } x \in R^2: \quad \varphi{(x_1, x_2)} = (1, x_1, \ x_2, \ x_1x_2, \ x_1^2, \ x_2^2, \ x_1^2x_2, \ x_1x_2^2, \ x_1^3, \ x_2^3)$
### Covers Theorem 

$\boxed{\color{orchid}\text{"A complex pattern classification problem cast in a high dimensional space nonlinearly is increasingly more likely to be linearly seperable" }}$

For the proof follow : https://web.mit.edu/course/other/i2course/www/vision_and_learning/perceptron_notes.pdf

### Dimensional Explosion 

$\text{There's a single major problem with the current solution: the full rational (of degree r) variety will increase the dimension from }$ $ \large{p \rightarrow \binom{p + r}{r} }$
