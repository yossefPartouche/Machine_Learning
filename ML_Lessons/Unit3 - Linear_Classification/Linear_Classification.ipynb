# Linear Classification I

$\text{The classification problem: Given an unordered list of class/labels, we wish to correctly predict that class for a set of sample points}$

## Discriminant Function

- $\text{Given a datset that has a binary classification (meaning can have only two labels/classes)}$
- $\text{Our sample points have p-features}$
- $\text{If our classifier is binary we can always just convert to \{1,0\} or \{1, -1\} depending what's more helpful.}$
- $\text{There's always a region in the hyperplane defined by the number of features,} \\  \text{which can be split into a region, M where the sample points are separated by their classification} \\ \text{The descision boundary is the line that seperate this region defined by the derivative M}$
- $\text{This region is define by a continouous function, f:}$ $$c(x) = \begin{cases} \ \ \ 1 \quad \text{if }  f(x) \ge 0 \\ -1 \quad \text{if } f(x) \lt 0\end{cases}$$
$$\quad \quad M = \{x \in \R^p \mid f(x) \ge 0 \} $$
$$ \quad \ \ \ \  \partial M = \{x \in \R^p | f(x) = 0 \}$$
- $\text{The } \partial \text{ notation is just a notation for a boundary set of points and does not refer to any derivation}$


$\text{Example}$
- $\text{In the image below we see the above describe over two features (the "x" and "y" axis)}$
    - $\text{The feature space is 2D}$
    - $\text{The hyperplane dimension: 1D (line)}$
    - $\text{The descision boundary is a 1D curve}$
    - $\text{The classification points are 0D (points)}$

## Linear Discriminant Functions 

$\color{dodgerblue}\boxed{\text{Linear Discriminant :=} f_w(\vec{x}) = w_0 + \sum_{i=1}^p w_ix_i \mid \ w \ , x \in \R^p}$

$\color{dodgerblue}\boxed{\text{Descision boundary := } \partial M_{f_w}  \\ = \{x \in \R^p \mid f_w(x) = 0\}}$

#### Example when $p = 2$
1. $f_w(x) = w_0 + w_1x_1 + w_2x_2 \mid w \in \R^3$
2. $\partial M_{f_w} = \{x \in \R^2 \mid f_w(x) = 0 \} \\= \{x \in \R^2 \mid w_0 + w_1x_1 + w_2x_2 =0 \}  \\ = \{(x_1, x_2) \in \R^2 \mid x_2 = -\frac{w_1}{w_2}x_1 - \frac{w_0}{w_2}\} $
- $\text{So positive classification region is: } M_{f_w} = \{x \in \R^2 \mid f_w > 0\} \text{ and this depends on the values of } w_1 \ , w_2 \text{ and } w_0.$

## Linear Seperability 

$\text{A dataset } D = \{(x_i, y_i)\}^n_{i=1} | \ \ x_i \in \R^p \ \ y^{(i)} \in \{-1, 1\} \\ \text{ is "linearly seperable" if there exists a weight vector } (w_0, \dots, w_p) \in \R^{p+1} \mid \ \forall (x,y) \in D \text{ we have } f_w(x) \gt 0 \iff y \gt 0$

#### Our Goal 

1. $\color{green}\text{Given a Dataset, determine if its linearly seperable}$
    - $\color{green}\text{If it is linearly seperable find the linear (discriminant) function}$

$\text{Assuming we found our predicting classifier (discriminant) function, meaning we found valid values for the vector } \vec{w} \\ \text{Then given a sample point } x^{(i)} \text{ we'll predict that class by the following function:}$ $$z_w(x^{(i)}) = sign(f_w(x^{(i)}))$$ 

#### Performance Evaluation

- $\large\text{Loss Function: } \frac{1}{2}|sign(w^{\top}x^{(i)}) - y^{(i)}|$
- $\large\text{Emperical Risk: } \frac{1}{n}\sum^n_{i=1}\frac{1}{2}|sign(w^{\top}x^{(i)}) - y^{(i)}|$

#### Our Goal (Formally)
1. $\color{green}\large \argmin_{w \in \R^{p+1}} \{\frac{1}{n}\sum^n_{i=1}\frac{1}{2}|sign(w^{\top}x^{(i)}) - y^{(i)}|\}$
    - $\color{green}\sum^n_{i=1}\frac{1}{2}|sign(w^{\top}x^{(i)}) - y^{(i)}| = 0$

## The Perceptron Algorithm 

$\text{A Dataset } D = \{(x_i, y_i)\}^n_{i=1} | x_i \in \R^p+1 y_i \in \{-1, 1\}$
- $\text{Here we set } x_0 = 1 \text{ for the convience of notation and implmentation}$


**Algorithm:**

$
\begin{aligned}
&\text{1. Start with a random } \mathbf{w} \in \mathbb{R}^{p+1}. \\
&\text{2. Repeat until convergence:} \\
&\quad \text{2.1. For each training sample } (\mathbf{x}^{(i)}, y^{(i)}) \in \mathcal{D}: \\
&\qquad \text{2.2. If } \text{sign}(\mathbf{w}^\top \mathbf{x}^{(i)}) \neq y^{(i)}, \text{ adjust } \mathbf{w}: \\
&\qquad \mathbf{w} \leftarrow \mathbf{w} + \Delta \mathbf{w}, \text{ where } \\
&\qquad \Delta \mathbf{w} = -\eta \big( \text{sign}(\mathbf{w}^\top \mathbf{x}^{(i)}) - y^{(i)} \big) \mathbf{x}^{(i)}.
\end{aligned}
$

#### Adjusting 
$\text{The peicewise function below presents the possible values that we can obtain when comparing the prediction with actual value}$

$sign(w^{\top}x^{(i)}) - y^{(i)}\begin{cases} \ \ \  0 \text{ when } z_w(x^{(i)}) = y^{(i)} \\ \ \ \ 2 \text{ when } 1 = z_w(x^{(i)}) = -y^{(i)} \\ -2 \text{ when } -1 = z_w(x^{(i)}) = -y^{(i)}\end{cases}$

$\text{As such adding the - minus will naturally update the value in the opposite direction from where it deviated (i.e. trying to bring it closer to 0)} \\ \text{and } \eta \text{ is indeed the same learning rate term as we saw in the Gradient descent algorithm}$

#### Remarks

1. $\text{If the data are } \bold{linearly \ seperable} \text{ and } \eta \gt 0 \text{ is small enough, then the algorithm } \bold{guarrantees } \text{ to converge to a seperating hyperplane}$
2. $\text{The number of steps required till we reach converge point, can be very large}$
3. $\text{If the data isn't linearly seperable or } \eta \text{ is too large, the algorithm will not converge and instead will oscillate, which can take a while for us to detect}$
4. $\text{Although this algorithm isn't the most pactical learning algorithm it's an important method to understand that's used in } \bold{Artificial \ Neural \ Networks}$

## Least Mean Squares

| Set up | Formalised|
|-------|------------|
| $$\text{Data}$$ |$$D = \{(x_i, y_i)\}^n_{i=1} \mid \ \ x_i \in \R^p \ \ y^{(i)} \in \{-1, 1\} $$ |
| $$\text{Loss function}$$ | $$ J(\vec{w}; D) = \sum^n_{i=1}(\vec{w}^{\top}x^{(i)} - y^{(i)})^2$$|
| $$w^*$$ | $$\argmin_{w \in \R^{p+1}} J(\vec{w}; D)$$|
| $$\text{Analytically}$$ | $$w^* = (X^{\top}X)^{-1}X^{\top}y$$| 
| $$\text{Iteratively}$$ | $$\vec{w_{update}} = \vec{w} - \eta \nabla J(\vec{w})$$|
| $$\text{To predict (Classify)}$$ | $$z_{w^*}(x) = sign(\vec{w}^{*\top}x)$$|

</center>

<center>

#### Algorithm Variations

| $$\textbf{ LMS}(\mathcal{D}, \eta)$$ | $$\textbf{Stochastic LMS}(\mathcal{D}, \eta)$$| $$\textbf{Mini-Batch LMS}(\mathcal{D}, \eta)$$ | 
| - |- | - | 
|$ \begin{aligned} &\text{1. Set t = 0 and start with a random } \mathbf{w^{(0)}} \in \mathbb{R}^{p+1}. \\ &\text{2. Compute the gradient} \nabla J(w) \text{ at } w^{(t)} \\ &\text{3. Update } w \text{ in opposite direction of gradient:} \\  &\quad \quad w^{(t+1)} = w^{(t)} - \eta \nabla J (w^{(t)})\\ &\text{4. If not converged, then t = t+1 and go to step (2)} \end{aligned} $ |$ \begin{aligned}  &\text{1. Set } t = 0 \text{ and start with a random } \mathbf{w^{(0)}} \in \mathbb{R}^{p+1}. \\  &\text{2. For each training example } (x^{(i)}, y^{(i)}): \\  &\quad \text{a. Predict: } \hat{y}^{(i)} = (w^{(t)})^\top x^{(i)} \\  &\quad \text{b. Compute the error: } e^{(i)} = \hat{y}^{(i)} - y^{(i)} \\  &\quad \text{c. Update: } w^{(t+1)} = w^{(t)} - \eta \cdot e^{(i)} x^{(i)} \\ &\text{3. If not converged, repeat for next sample (or next epoch).} \end{aligned} $ | $\begin{aligned}&\text{1. Initialize } \mathbf{w} \in \mathbb{R}^{p+1} \text{ randomly} \\ &\text{2. Repeat until convergence:} \\ &\quad \bullet \text{ Select a random mini-batch } S \subseteq \{1, 2, \dots, n\},\ \mid S \mid = m \\ &\quad \bullet \text{Compute the average gradient over the mini-batch:} \\ &\quad \quad \Delta \mathbf{w} = -\frac{\eta}{ \mid S \mid } \sum_{i \in S} \left( \mathbf{w}^\top \mathbf{x}^{(i)} - y^{(i)} \right) \mathbf{x}^{(i)} \\ &\quad \bullet \text{Update weights:} \\ &\quad \quad \mathbf{w} \leftarrow \mathbf{w} + \Delta \mathbf{w} \end{aligned}$|

</center>

$$ \boxed{\nabla J(w) = 2 \sum^n_{i=1}\left(w^{\top}x^{(i)} - y^{(i)}\right)x^{(i)}}$$

### Goal of LMS

- $\text{This algorithm find the parameters which } \bold{minimizes \ the \ sum \ of \ the \ squares \ distances} \text{ between the training samples and the } \bold{offset} \text{ +1, or -1 to the hyperplane}$
    - $\text{Meaning when we compute the loss: }\sum^n_{i=1}(\vec{w}^{\top}x^{(i)} - y^{(i)})^2 \text{ we're actually copmuting: } \sum_{(x,y) \in D^+}(\vec{w}^{\top}x^{(i)} - 1)^2 + \sum_{(x,y) \in D^-}(\vec{w}^{\top}x^{(i)} + 1)^2$ 
- $\text{Contrast to the perceptron algorithm which counts the number of misclassification, this algorithm reduces the distance}$
- $\text{In consequence this could lead to data samples being misclassified}$
