#### Question 1
$$ l(\lambda; D) :~=~ log[l(\lambda; D)] = \sum_{i=1}^n[p_{\lambda}(x_i)] = \sum_{i=1}^n\log(\frac{\lambda^{x_i}e^{-\lambda}}{x_i!})$$

$$ \sum_{i=1}^n\log(\frac{\lambda^{x_i}e^{-\lambda}}{x_i!}) = \sum_{i=1}^n[\log(\lambda^{x_i}) + \log(e^{-\lambda}) - \log(x_i!)]$$

$$ = \sum_{i=1}^n[x_i\log(\lambda) - \lambda - \log(x_i!)]$$

$$ = -\lambda n + \log(\lambda)\sum_{i=1}^n x_i - \sum_{i=1}^n\log(x_i!)$$

#### Question 2

To find the MLE $\hat{\lambda}$, we need to maximize the log-likelihood function from Question 1:

$$ l(\lambda; D) = -\lambda n + \log(\lambda)\sum_{i=1}^n x_i - \sum_{i=1}^n\log(x_i!) $$

$$ \frac{d}{d\lambda}l(\lambda; D) = -n + \frac{1}{\lambda}\sum_{i=1}^n x_i = 0 $$

Solving for $\lambda$:

$$ \frac{1}{\lambda}\sum_{i=1}^n x_i = n $$

$$ \sum_{i=1}^n x_i = n\lambda $$

$$ \hat{\lambda} = \frac{1}{n}\sum_{i=1}^n x_i $$

#### Question 3
Starting with the derived MLE for the Poisson parameter:

$$\hat{\lambda} = \frac{1}{n}\sum_{i=1}^n x_i$$

Where $x_i \sim \text{Poisson}(\lambda)$ are independent samples.

Using the linearity of expectation:

$$\mathbb{E}[\hat{\lambda}] = \mathbb{E}\left[\frac{1}{n}\sum_{i=1}^n x_i\right] = \frac{1}{n}\sum_{i=1}^n \mathbb{E}[x_i]$$

Since each $x_i$ follows a Poisson distribution with parameter $\lambda$, we know that $\mathbb{E}[x_i] = \lambda$. Therefore:

$$\mathbb{E}[\hat{\lambda}] = \frac{1}{n}\sum_{i=1}^n \lambda = \frac{n\lambda}{n} = \lambda$$

For the variance, we use the property that the variance of a sum of independent random variables equals the sum of their individual variances:

$$\text{Var}[\hat{\lambda}] = \text{Var}\left[\frac{1}{n}\sum_{i=1}^n x_i\right] = \frac{1}{n^2}\sum_{i=1}^n \text{Var}[x_i]$$

For a Poisson random variable, $\text{Var}[x_i] = \lambda$. Therefore:

$$\text{Var}[\hat{\lambda}] = \frac{1}{n^2}\sum_{i=1}^n \lambda = \frac{1}{n^2} \cdot n\lambda = \frac{\lambda}{n}$$

Thus, the variance of the MLE decreases proportionally to the sample size $n$.
