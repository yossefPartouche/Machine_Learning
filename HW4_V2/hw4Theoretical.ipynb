#### 1. Log-Likelihood Function

$\text{Assuming samples \((x^{(i)}, y^{(i)})\) are i.i.d. and that \(P_X(x^{(i)})\) does not depend on \(w\), the log-likelihood is:}$

$
\ell(w; D) = \sum_{i=1}^{n} \log \left[ P(x^{(i)}, y^{(i)} \mid w) \right]
= \sum_{i=1}^{n} \log \left[ P(x^{(i)} \mid w) \cdot P(y^{(i)} \mid x^{(i)}, w) \right]
$

$\text{Ignoring } P(x^{(i)} \mid w) \text{ as it is independent of } w, \text{ we get:}$


$
\ell(w; D) = \sum_{i=1}^{n} \log \left[ P(y^{(i)} \mid x^{(i)}, w) \right]
$

$\text{The class-conditional probability is:}$

$
P(y^{(i)} = 1 \mid x^{(i)}, w) = \Phi(w^\top x^{(i)})
$

$
P(y^{(i)} = 0 \mid x^{(i)}, w) = 1 - \Phi(w^\top x^{(i)})
$

So the log-likelihood becomes:

$
\ell(w; D) = \sum_{i=1}^{n} \left[ y^{(i)} \log \Phi(w^\top x^{(i)}) + (1 - y^{(i)}) \log(1 - \Phi(w^\top x^{(i)})) \right]
$

### 2. Binary Cross-Entropy Loss

Maximizing the log-likelihood is equivalent to minimizing the binary cross-entropy (BCE) loss:

$$
\mathcal{L}(w) = - \sum_{i=1}^{n} \left[ y^{(i)} \log \Phi(w^\top x^{(i)}) + (1 - y^{(i)}) \log(1 - \Phi(w^\top x^{(i)})) \right]
$$
---

### 3. Gradient of the BCE Loss

Recall that:

$$
\Phi'(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2 / 2}
$$

Using the chain rule, the gradient of the BCE loss is:

$$
\nabla \mathcal{L}(w) = - \sum_{i=1}^{n} \left[ \left( \frac{y^{(i)}}{\Phi(w^\top x^{(i)})} - \frac{1 - y^{(i)}}{1 - \Phi(w^\top x^{(i)})} \right) \cdot \Phi'(w^\top x^{(i)}) \cdot x^{(i)} \right]
$$

Substituting for $\Phi'(w^\top x^{(i)})$:

$$
\nabla \mathcal{L}(w) = - \sum_{i=1}^{n} \left[ \left( \frac{y^{(i)}}{\Phi(w^\top x^{(i)})} - \frac{1 - y^{(i)}}{1 - \Phi(w^\top x^{(i)})} \right) \cdot \frac{1}{\sqrt{2\pi}} e^{-(w^\top x^{(i)})^2 / 2} \cdot x^{(i)} \right]
$$
